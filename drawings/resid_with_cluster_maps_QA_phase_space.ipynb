```python
input_file = root.TFile(file_path+file_name)
if not input_file or input_file.IsZombie():
    raise RuntimeError(f"Could not open ROOT file: {file_path+file_name}")

hists = []
missing_histograms = []

for idir in Directories:
    h_folder = []
    for hist_name in hist_names:
        h = input_file.Get(idir+"/"+hist_name)
        if not h:
            missing_histograms.append(idir+"/"+hist_name)
            h_folder.append(None)
            continue
        h.SetDirectory(0)
        h.SetName(hist_name+"_"+idir)
        h_folder.append(h)
    hists.append(h_folder)

input_file.Close()

if missing_histograms:
    print("WARNING: the following histograms were not found:")
    for name in missing_histograms:
        print("  ", name)
```

```python
#going form hist array to hist dictionary
hists_dict = {}
for i, idir in enumerate(Directories):
    hists_dict[idir] = {}
    for j, hist_name in enumerate(hist_names):
        hists_dict[idir][hist_name] = hists[i][j]
```

```python
def Format_hist_loc(h, title, x_title, y_title, x_title_size=0.06, y_title_size=0.06, x_title_offset=1.0, \
                    y_title_offset=1.0, label_size_x=0.06, label_size_y=0.06, label_size_z=0.06,\
                         z_axis_offset=1.0, z_range = None, center_title=False, title_size=0.06):
    if h is None:
        print("Error: Histogram is None. Cannot format.")
        return
    if title is not None:
        h.SetTitle(title)
    if x_title is not None:
        h.GetXaxis().SetTitle(x_title)
    if y_title is not None:
        h.GetYaxis().SetTitle(y_title)
    h.GetXaxis().SetTitleSize(x_title_size)
    h.GetYaxis().SetTitleSize(y_title_size)
    h.GetXaxis().SetTitleOffset(x_title_offset)
    h.GetYaxis().SetTitleOffset(y_title_offset)
    h.GetXaxis().SetLabelSize(label_size_x)
    h.GetYaxis().SetLabelSize(label_size_y)
    h.GetZaxis().SetLabelSize(label_size_z)
    h.GetZaxis().SetTitleOffset(z_axis_offset)
    #h.SetTitleSize(title_size)
    if z_range is not None:
        h.GetZaxis().SetRangeUser(z_range[0], z_range[1])
    if center_title:
        h.GetXaxis().CenterTitle()
        h.GetYaxis().CenterTitle()
```

```python

#helper functions for layer/phi and signed-z/layer residual maps
#keeps the same ROOT drawing style used above

def Draw_layer_lines(xmin, xmax, ymin, ymax, x_is_layer=False, y_is_layer=False):
    lines = []
    for layer_edge in [7, 23, 39, 55]:
        if x_is_layer:
            line = root.TLine(layer_edge, ymin, layer_edge, ymax)
            line.SetLineColor(root.kBlack)
            line.SetLineWidth(1)
            line.SetLineStyle(2)
            line.Draw("same")
            lines.append(line)
        if y_is_layer:
            line = root.TLine(xmin, layer_edge, xmax, layer_edge)
            line.SetLineColor(root.kBlack)
            line.SetLineWidth(1)
            line.SetLineStyle(2)
            line.Draw("same")
            lines.append(line)
    return lines

def Draw_phi_lines(xmin, xmax):
    lines = []
    sector_angle = root.TMath.Pi()/6.
    angle_start = root.TMath.Pi() - sector_angle/2.
    for isector in range(12):
        phi_edge = angle_start - isector*sector_angle
        if phi_edge > root.TMath.Pi():
            phi_edge -= 2.*root.TMath.Pi()
        if phi_edge < -root.TMath.Pi():
            phi_edge += 2.*root.TMath.Pi()
        line = root.TLine(xmin, phi_edge, xmax, phi_edge)
        line.SetLineColor(root.kBlack)
        line.SetLineWidth(1)
        line.SetLineStyle(2)
        line.Draw("same")
        lines.append(line)
    return lines

def Draw_z_lines(ymin, ymax):
    lines = []
    for z_edge in [-102.325, 0., 102.325]:
        line = root.TLine(z_edge, ymin, z_edge, ymax)
        line.SetLineColor(root.kBlack)
        line.SetLineWidth(1)
        line.SetLineStyle(2)
        line.Draw("same")
        lines.append(line)
    return lines

def Format_residual_map(h, title, x_title, y_title, z_title, x_range, y_range, z_range):
    Format_hist_loc(h, title=title, x_title=x_title, y_title=y_title,
                    x_title_offset=.9, y_title_offset=1.0,
                    center_title=True, x_title_size=0.06,
                    y_title_size=0.06, label_size_x=0.05,
                    label_size_y=0.05, label_size_z=0.05,
                    z_axis_offset=1.15, z_range=z_range)
    h.GetZaxis().SetTitle(z_title)
    h.GetXaxis().SetRangeUser(x_range[0], x_range[1])
    h.GetYaxis().SetRangeUser(y_range[0], y_range[1])
```

```python
#make rDCAzero vs pT projections from h_rDCA_vs_qOverPt_vs_pt
#
#original TH3 axes:
#X = q/pT
#Y = rDCAzero
#Z = pT
#
#pT binning:
#0.2 GeV/c bins from 0.2 to 3.0 GeV/c
#0.5 GeV/c bins from 3.0 to 5.0 GeV/c
#
#profile option "s" stores the standard deviation as the profile-bin error

from array import array
import math

hist_name = "h_rDCA_vs_qOverPt_vs_pt"

dirs_dca_pt = [
    'side0_qplus',
    'side0_qminus',
    'side1_qplus',
    'side1_qminus'
]

names_dca_pt = [
    'side0 positive',
    'side0 negative',
    'side1 positive',
    'side1 negative'
]

colors_dca_pt = [
    root.kRed+1,
    root.kBlue+1,
    root.kMagenta+1,
    root.kGreen+2
]

markers_dca_pt = [20, 21, 24, 25]

pt_edges = []

pt_value = 0.2
while pt_value < 3.0-1.e-8:
    pt_edges.append(pt_value)
    pt_value += 0.2

pt_edges.append(3.0)
pt_edges += [3.5, 4.0, 4.5, 5.0]

pt_edges_array = array('d', pt_edges)

dca_pt_2d = []
dca_pt_profiles = []
dca_pt_means = []
dca_pt_sigmas = []

for icase, idir in enumerate(dirs_dca_pt):

    h3 = hists_dict[idir][hist_name]

    #keep the original rDCAzero binning, restricted to -9 < rDCAzero < 9 cm
    first_dca_bin = h3.GetYaxis().FindBin(-9. + 1.e-6)
    last_dca_bin  = h3.GetYaxis().FindBin( 9. - 1.e-6)

    n_dca_bins = last_dca_bin-first_dca_bin+1

    dca_low = h3.GetYaxis().GetBinLowEdge(first_dca_bin)
    dca_high = h3.GetYaxis().GetBinUpEdge(last_dca_bin)

    h2 = root.TH2D(
        f"h_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array,
        n_dca_bins,
        dca_low,
        dca_high
    )

    h2.SetDirectory(0)
    h2.Sumw2()

    #sum over all q/pT bins and rebin pT into the requested variable bins
    for iz in range(1, h3.GetNbinsZ()+1):

        pt = h3.GetZaxis().GetBinCenter(iz)

        if pt < pt_edges[0] or pt >= pt_edges[-1]:
            continue

        new_pt_bin = h2.GetXaxis().FindBin(pt)

        for iy in range(first_dca_bin, last_dca_bin+1):

            rdca = h3.GetYaxis().GetBinCenter(iy)

            content_sum = 0.
            error2_sum = 0.

            for ix in range(1, h3.GetNbinsX()+1):

                content = h3.GetBinContent(ix, iy, iz)
                error = h3.GetBinError(ix, iy, iz)

                content_sum += content
                error2_sum += error*error

            new_dca_bin = h2.GetYaxis().FindBin(rdca)

            old_content = h2.GetBinContent(new_pt_bin, new_dca_bin)
            old_error = h2.GetBinError(new_pt_bin, new_dca_bin)

            h2.SetBinContent(
                new_pt_bin,
                new_dca_bin,
                old_content+content_sum
            )

            h2.SetBinError(
                new_pt_bin,
                new_dca_bin,
                math.sqrt(old_error*old_error+error2_sum)
            )

    #profile mean rDCAzero versus pT
    #option "s": profile errors are standard deviations
    profile = h2.ProfileX(
        f"p_rDCAzero_vs_pt_{idir}",
        1,
        h2.GetNbinsY(),
        "s"
    )

    profile.SetDirectory(0)

    #separate mean histogram
    #mean uncertainty = sigma/sqrt(N)
    hmean = root.TH1D(
        f"h_mean_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array
    )

    hmean.SetDirectory(0)

    #separate sigma histogram
    hsigma = root.TH1D(
        f"h_sigma_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array
    )

    hsigma.SetDirectory(0)

    for ibin in range(1, profile.GetNbinsX()+1):

        mean = profile.GetBinContent(ibin)
        sigma = profile.GetBinError(ibin)
        entries = profile.GetBinEntries(ibin)

        mean_error = sigma/math.sqrt(entries) if entries > 0 else 0.

        hmean.SetBinContent(ibin, mean)
        hmean.SetBinError(ibin, mean_error)

        hsigma.SetBinContent(ibin, sigma)
        hsigma.SetBinError(ibin, 0.)

    h2.SetLineColor(colors_dca_pt[icase])
    profile.SetLineColor(colors_dca_pt[icase])
    profile.SetMarkerColor(colors_dca_pt[icase])
    profile.SetMarkerStyle(markers_dca_pt[icase])

    hmean.SetLineColor(colors_dca_pt[icase])
    hmean.SetLineWidth(2)
    hmean.SetMarkerColor(colors_dca_pt[icase])
    hmean.SetMarkerStyle(markers_dca_pt[icase])
    hmean.SetMarkerSize(1.0)

    hsigma.SetLineColor(colors_dca_pt[icase])
    hsigma.SetLineWidth(2)
    hsigma.SetMarkerColor(colors_dca_pt[icase])
    hsigma.SetMarkerStyle(markers_dca_pt[icase])
    hsigma.SetMarkerSize(1.0)

    dca_pt_2d.append(h2)
    dca_pt_profiles.append(profile)
    dca_pt_means.append(hmean)
    dca_pt_sigmas.append(hsigma)
```

```python
# Helpers for the new phase-space QA

def clone_rebin_2d(h, name, rebin_x=1, rebin_y=1):
    if h is None:
        return None
    out = h.Clone(name)
    out.SetDirectory(0)
    if rebin_x > 1 or rebin_y > 1:
        out.Rebin2D(int(rebin_x), int(rebin_y))
    return out

def clone_rebin_3d(h, name, rebin_x=1, rebin_y=1, rebin_z=1):
    if h is None:
        return None
    out = h.Clone(name)
    out.SetDirectory(0)
    if rebin_x > 1 or rebin_y > 1 or rebin_z > 1:
        out.Rebin3D(int(rebin_x), int(rebin_y), int(rebin_z))
    return out

def set_axis_range(h, axis, limits):
    if h is None or limits is None:
        return
    getattr(h, f"Get{axis.upper()}axis")().SetRangeUser(*limits)

def make_ratio_map(num, den, name):
    ratio = num.Clone(name)
    ratio.SetDirectory(0)
    ratio.Divide(den)
    return ratio

def draw_four_qa_maps(hist_name, x_title, y_title, rebin_x, rebin_y,
                      x_range=None, y_range=None, logz=False, z_range=None):
    canvas = root.TCanvas("c_"+hist_name, "c_"+hist_name, 1400, 900)
    canvas.Divide(2, 2)
    drawn = []

    for i, idir in enumerate(qa_dirs):
        canvas.cd(i+1)
        root.gPad.SetLeftMargin(0.12)
        root.gPad.SetBottomMargin(0.12)
        root.gPad.SetRightMargin(0.16)
        root.gPad.SetTopMargin(0.08)
        root.gPad.SetLogz(logz)

        h = clone_rebin_2d(
            hists_dict[idir][hist_name],
            f"{hist_name}_{idir}_coarse",
            rebin_x,
            rebin_y
        )
        if h is None:
            continue

        set_axis_range(h, "x", x_range)
        set_axis_range(h, "y", y_range)
        if z_range is not None:
            h.SetMinimum(z_range[0])
            h.SetMaximum(z_range[1])

        Format_hist_loc(
            h,
            title=f"{qa_labels[i]}, {data_set}, {qa_cuts}",
            x_title=x_title,
            y_title=y_title,
            x_title_offset=0.95,
            y_title_offset=0.9,
            center_title=True
        )
        h.Draw("COLZ")
        drawn.append(h)

    canvas.Update()
    return canvas, drawn
```

```python
# ============================================================
# Positive / negative yield ratios in selected pT intervals
#
# For every pT interval:
#   1. 2D positive/negative ratio map versus phi and eta
#   2. Positive/negative ratio versus phi, integrated over eta
#   3. Positive/negative ratio versus eta, integrated over phi
#
# Input TH3 axes:
#   X = phi
#   Y = eta
#   Z = pT
# ============================================================

import math

# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

pt_ratio_bins = [
    (0.8, 1.2),
    (1.2, 1.6),
    (1.6, 2.4),
    (2.4, 5.2),
]

# Your TH3 already has:
#   128 phi bins
#   100 eta bins
#   50 pT bins
#
# These rebin values make the maps substantially coarser.
ratio_map_phi_rebin = 8
ratio_map_eta_rebin = 5

# Additional rebinning for the 1D projections.
# The TH3 is already rebinned before projection.
ratio_projection_phi_rebin = 1
ratio_projection_eta_rebin = 1

eta_draw_range = (-1.2, 1.2)

ratio_1d_range = (0.7, 1.3)
ratio_2d_range = (0.7, 1.3)

minimum_denominator_entries = 1.0

pt_phi_eta_hist_name = "h_pt_vs_atan2_px_py_vs_eta_rdca3"

side_pairs = [
    ("side0_qplus", "side0_qminus", "side0"),
    ("side1_qplus", "side1_qminus", "side1"),
]

pt_colors = [
    root.kBlack,
    root.kRed + 1,
    root.kBlue + 1,
    root.kMagenta + 1,
    root.kGreen + 2,
    root.kOrange + 7,
]

pt_markers = [20, 21, 22, 23, 33, 34]


# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------

def clone_and_rebin_phase_space(
    h3,
    name,
    phi_rebin,
    eta_rebin
):
    """
    Clone TH3 and rebin only phi and eta.

    The pT axis is intentionally left unchanged.
    """

    result = h3.Clone(name)
    result.SetDirectory(0)
    result.Sumw2()

    if phi_rebin > 1 or eta_rebin > 1:
        result.Rebin3D(
            phi_rebin,
            eta_rebin,
            1
        )

    return result


def pt_bin_range(h3, pt_low, pt_high):
    """
    Return inclusive ROOT Z-bin indices for a pT interval.
    """

    first_bin = h3.GetZaxis().FindBin(pt_low + 1.0e-6)
    last_bin = h3.GetZaxis().FindBin(pt_high - 1.0e-6)

    return first_bin, last_bin


def make_phi_eta_yield_map(
    h3,
    name,
    pt_low,
    pt_high
):
    """
    Project a TH3 into a TH2(phi, eta) yield map
    for one selected pT interval.
    """

    first_pt_bin, last_pt_bin = pt_bin_range(
        h3,
        pt_low,
        pt_high
    )

    # Project3D("yx") gives:
    #   X axis = original X = phi
    #   Y axis = original Y = eta
    #
    # ROOT naming can be confusing, so verify the axis titles
    # when first running this.
    old_first = h3.GetZaxis().GetFirst()
    old_last = h3.GetZaxis().GetLast()

    h3.GetZaxis().SetRange(
        first_pt_bin,
        last_pt_bin
    )

    result = h3.Project3D("yx")
    result.SetName(name)
    result.SetDirectory(0)
    result.Sumw2()

    h3.GetZaxis().SetRange(
        old_first,
        old_last
    )

    return result


def make_safe_ratio_2d(
    numerator,
    denominator,
    name
):
    """
    Create numerator/denominator with explicit empty-bin handling.
    """

    ratio = numerator.Clone(name)
    ratio.SetDirectory(0)
    ratio.Reset("ICES")
    ratio.Sumw2()

    for ix in range(1, ratio.GetNbinsX() + 1):
        for iy in range(1, ratio.GetNbinsY() + 1):

            num = numerator.GetBinContent(ix, iy)
            den = denominator.GetBinContent(ix, iy)

            num_error = numerator.GetBinError(ix, iy)
            den_error = denominator.GetBinError(ix, iy)

            if den < minimum_denominator_entries:
                continue

            value = num / den

            error_squared = (
                (num_error / den) ** 2
                + (num * den_error / (den * den)) ** 2
            )

            ratio.SetBinContent(ix, iy, value)
            ratio.SetBinError(
                ix,
                iy,
                math.sqrt(max(0.0, error_squared))
            )

    ratio._numerator = numerator
    ratio._denominator = denominator

    return ratio


def make_safe_ratio_1d(
    numerator,
    denominator,
    name
):
    """
    Create a statistically correct 1D numerator/denominator ratio.
    """

    ratio = numerator.Clone(name)
    ratio.SetDirectory(0)
    ratio.Reset("ICES")
    ratio.Sumw2()

    for ibin in range(1, ratio.GetNbinsX() + 1):

        num = numerator.GetBinContent(ibin)
        den = denominator.GetBinContent(ibin)

        num_error = numerator.GetBinError(ibin)
        den_error = denominator.GetBinError(ibin)

        if den < minimum_denominator_entries:
            continue

        value = num / den

        error_squared = (
            (num_error / den) ** 2
            + (num * den_error / (den * den)) ** 2
        )

        ratio.SetBinContent(ibin, value)
        ratio.SetBinError(
            ibin,
            math.sqrt(max(0.0, error_squared))
        )

    ratio._numerator = numerator
    ratio._denominator = denominator

    return ratio


def project_yield_ratio_to_phi(
    plus_map,
    minus_map,
    name
):
    """
    Integrate positive and negative yields over eta first,
    then divide.

    This is preferable to projecting an already-divided ratio map.
    """

    plus_phi = plus_map.ProjectionX(
        name + "_plus",
        1,
        plus_map.GetNbinsY(),
        "e"
    )

    minus_phi = minus_map.ProjectionX(
        name + "_minus",
        1,
        minus_map.GetNbinsY(),
        "e"
    )

    plus_phi.SetDirectory(0)
    minus_phi.SetDirectory(0)

    if ratio_projection_phi_rebin > 1:
        plus_phi.Rebin(ratio_projection_phi_rebin)
        minus_phi.Rebin(ratio_projection_phi_rebin)

    return make_safe_ratio_1d(
        plus_phi,
        minus_phi,
        name
    )


def project_yield_ratio_to_eta(
    plus_map,
    minus_map,
    name
):
    """
    Integrate positive and negative yields over phi first,
    then divide.
    """

    plus_eta = plus_map.ProjectionY(
        name + "_plus",
        1,
        plus_map.GetNbinsX(),
        "e"
    )

    minus_eta = minus_map.ProjectionY(
        name + "_minus",
        1,
        minus_map.GetNbinsX(),
        "e"
    )

    plus_eta.SetDirectory(0)
    minus_eta.SetDirectory(0)

    if ratio_projection_eta_rebin > 1:
        plus_eta.Rebin(ratio_projection_eta_rebin)
        minus_eta.Rebin(ratio_projection_eta_rebin)

    return make_safe_ratio_1d(
        plus_eta,
        minus_eta,
        name
    )


# ------------------------------------------------------------
# Build ratio maps and projections
# ------------------------------------------------------------

yield_ratio_maps = {}
yield_ratio_phi = {}
yield_ratio_eta = {}

yield_ratio_map_canvases = []
yield_ratio_projection_canvases = []

for ipt, (pt_low, pt_high) in enumerate(pt_ratio_bins):

    # ========================================================
    # 2D ratio maps
    # ========================================================

    canvas_map = root.TCanvas(
        f"c_yield_ratio_map_ptbin{ipt}",
        f"c_yield_ratio_map_ptbin{ipt}",
        1500,
        650
    )

    canvas_map.Divide(2, 1)

    keep_maps = []
    keep_input_maps = []

    for ipad, (plus_dir, minus_dir, side_name) in enumerate(side_pairs):

        hplus3 = clone_and_rebin_phase_space(
            hists_dict[plus_dir][pt_phi_eta_hist_name],
            f"hplus3_{side_name}_ptbin{ipt}",
            ratio_map_phi_rebin,
            ratio_map_eta_rebin
        )

        hminus3 = clone_and_rebin_phase_space(
            hists_dict[minus_dir][pt_phi_eta_hist_name],
            f"hminus3_{side_name}_ptbin{ipt}",
            ratio_map_phi_rebin,
            ratio_map_eta_rebin
        )

        plus_map = make_phi_eta_yield_map(
            hplus3,
            f"hplus_phi_eta_{side_name}_ptbin{ipt}",
            pt_low,
            pt_high
        )

        minus_map = make_phi_eta_yield_map(
            hminus3,
            f"hminus_phi_eta_{side_name}_ptbin{ipt}",
            pt_low,
            pt_high
        )

        ratio_map = make_safe_ratio_2d(
            plus_map,
            minus_map,
            f"h_yield_ratio_phi_eta_{side_name}_ptbin{ipt}"
        )

        yield_ratio_maps[
            (side_name, pt_low, pt_high)
        ] = ratio_map

        ratio_phi = project_yield_ratio_to_phi(
            plus_map,
            minus_map,
            f"h_yield_ratio_phi_{side_name}_ptbin{ipt}"
        )

        ratio_eta = project_yield_ratio_to_eta(
            plus_map,
            minus_map,
            f"h_yield_ratio_eta_{side_name}_ptbin{ipt}"
        )

        yield_ratio_phi[
            (side_name, pt_low, pt_high)
        ] = ratio_phi

        yield_ratio_eta[
            (side_name, pt_low, pt_high)
        ] = ratio_eta

        canvas_map.cd(ipad + 1)

        root.gPad.SetLeftMargin(0.12)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetRightMargin(0.17)
        root.gPad.SetTopMargin(0.10)

        Format_hist_loc(
            ratio_map,
            title=(
                f"{side_name}: positive / negative, "
                f"{pt_low:.1f} < p_{{T}} < {pt_high:.1f} GeV/c"
            ),
            x_title="#phi = atan2(p_{x},p_{y}) [rad]",
            y_title="#eta",
            x_title_offset=0.95,
            y_title_offset=0.95,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.052,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_map.GetYaxis().SetRangeUser(*eta_draw_range)

        ratio_map.GetZaxis().SetTitle(
            "positive / negative"
        )
        ratio_map.GetZaxis().CenterTitle()
        ratio_map.GetZaxis().SetTitleOffset(1.25)
        ratio_map.GetZaxis().SetTitleSize(0.045)
        ratio_map.GetZaxis().SetLabelSize(0.04)

        ratio_map.SetMinimum(ratio_2d_range[0])
        ratio_map.SetMaximum(ratio_2d_range[1])

        ratio_map.Draw("COLZ")

        keep_maps.append(ratio_map)
        keep_input_maps += [plus_map, minus_map, hplus3, hminus3]

    canvas_map.Update()
    canvas_map.Draw()

    canvas_map._maps = keep_maps
    canvas_map._inputs = keep_input_maps

    yield_ratio_map_canvases.append(canvas_map)

    if safe_to_pdf:
        canvas_map.Print(
            outdir
            + (
                f"/qa_yield_ratio_phi_eta_"
                f"pt_{pt_low:.1f}_{pt_high:.1f}.pdf"
            )
        )

    # ========================================================
    # Phi and eta projections
    # ========================================================

    canvas_projection = root.TCanvas(
        f"c_yield_ratio_projections_ptbin{ipt}",
        f"c_yield_ratio_projections_ptbin{ipt}",
        1500,
        1100
    )

    canvas_projection.Divide(2, 2)

    keep_projections = []
    keep_lines = []

    for iside, (_, _, side_name) in enumerate(side_pairs):

        ratio_phi = yield_ratio_phi[
            (side_name, pt_low, pt_high)
        ]

        ratio_eta = yield_ratio_eta[
            (side_name, pt_low, pt_high)
        ]

        ratio_phi.SetLineColor(root.kBlack)
        ratio_phi.SetMarkerColor(root.kBlack)
        ratio_phi.SetMarkerStyle(20)
        ratio_phi.SetMarkerSize(0.9)
        ratio_phi.SetLineWidth(2)

        ratio_eta.SetLineColor(root.kBlack)
        ratio_eta.SetMarkerColor(root.kBlack)
        ratio_eta.SetMarkerStyle(20)
        ratio_eta.SetMarkerSize(0.9)
        ratio_eta.SetLineWidth(2)

        # Phi projection
        canvas_projection.cd(2 * iside + 1)

        root.gPad.SetLeftMargin(0.13)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetGridy()

        Format_hist_loc(
            ratio_phi,
            title=(
                f"{side_name}: positive / negative, "
                f"{pt_low:.1f} < p_{{T}} < {pt_high:.1f} GeV/c"
            ),
            x_title="#phi = atan2(p_{x},p_{y}) [rad]",
            y_title="positive / negative",
            x_title_offset=0.95,
            y_title_offset=1.05,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.052,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_phi.SetMinimum(ratio_1d_range[0])
        ratio_phi.SetMaximum(ratio_1d_range[1])
        ratio_phi.Draw("E1")

        phi_line = root.TLine(
            -root.TMath.Pi(),
            1.0,
            root.TMath.Pi(),
            1.0
        )

        phi_line.SetLineStyle(2)
        phi_line.SetLineWidth(2)
        phi_line.SetLineColor(root.kGray + 2)
        phi_line.Draw("same")

        # Eta projection
        canvas_projection.cd(2 * iside + 2)

        root.gPad.SetLeftMargin(0.13)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetGridy()

        Format_hist_loc(
            ratio_eta,
            title=(
                f"{side_name}: positive / negative, "
                f"{pt_low:.1f} < p_{{T}} < {pt_high:.1f} GeV/c"
            ),
            x_title="#eta",
            y_title="positive / negative",
            x_title_offset=0.95,
            y_title_offset=1.05,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.052,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_eta.GetXaxis().SetRangeUser(*eta_draw_range)
        ratio_eta.SetMinimum(ratio_1d_range[0])
        ratio_eta.SetMaximum(ratio_1d_range[1])
        ratio_eta.Draw("E1")

        eta_line = root.TLine(
            eta_draw_range[0],
            1.0,
            eta_draw_range[1],
            1.0
        )

        eta_line.SetLineStyle(2)
        eta_line.SetLineWidth(2)
        eta_line.SetLineColor(root.kGray + 2)
        eta_line.Draw("same")

        keep_projections += [ratio_phi, ratio_eta]
        keep_lines += [phi_line, eta_line]

    canvas_projection.Update()
    canvas_projection.Draw()

    canvas_projection._projections = keep_projections
    canvas_projection._lines = keep_lines

    yield_ratio_projection_canvases.append(
        canvas_projection
    )

    if safe_to_pdf:
        canvas_projection.Print(
            outdir
            + (
                f"/qa_yield_ratio_phi_eta_projections_"
                f"pt_{pt_low:.1f}_{pt_high:.1f}.pdf"
            )
        )
```

```python
def fit_pt_spectrum_slope(
    spectrum,
    name,
    fit_config
):
    """
    Fit:

        ln[N(pT)] = intercept + slope*pT

    inside the pT range specified by fit_config.
    """

    pt_min = fit_config["pt_min"]
    pt_max = fit_config["pt_max"]

    minimum_content = fit_config["minimum_content"]
    minimum_points = fit_config["minimum_points"]

    x_values = []
    y_values = []
    x_errors = []
    y_errors = []

    for ibin in range(1, spectrum.GetNbinsX() + 1):

        pt_value = spectrum.GetXaxis().GetBinCenter(ibin)
        content = spectrum.GetBinContent(ibin)
        content_error = spectrum.GetBinError(ibin)

        if pt_value < pt_min or pt_value > pt_max:
            continue

        if content < minimum_content:
            continue

        if content_error <= 0.0:
            content_error = math.sqrt(content)

        x_values.append(pt_value)
        y_values.append(math.log(content))
        x_errors.append(0.0)
        y_errors.append(content_error / content)

    if len(x_values) < minimum_points:
        return math.nan, math.nan, None, None

    graph = root.TGraphErrors(
        len(x_values),
        np.asarray(x_values, dtype="d"),
        np.asarray(y_values, dtype="d"),
        np.asarray(x_errors, dtype="d"),
        np.asarray(y_errors, dtype="d")
    )

    graph.SetName(name + "_graph")

    fit_function = root.TF1(
        name + "_fit",
        "[0]+[1]*x",
        pt_min,
        pt_max
    )

    fit_result = graph.Fit(
        fit_function,
        "QSN"
    )

    if int(fit_result) != 0:
        return math.nan, math.nan, graph, fit_function

    slope = fit_function.GetParameter(1)
    slope_error = fit_function.GetParError(1)

    # Reject failed or numerically meaningless results.
    if not math.isfinite(slope) or not math.isfinite(slope_error):
        return math.nan, math.nan, graph, fit_function

    return slope, slope_error, graph, fit_function
```

```python

# ------------------------------------------------------------
# TH3 preparation
# ------------------------------------------------------------

def prepare_slope_source(
    h3,
    name,
    fit_config
):
    source = h3.Clone(name)
    source.SetDirectory(0)
    source.Sumw2()

    source.Rebin3D(
        fit_config["phi_rebin"],
        fit_config["eta_rebin"],
        slope_map_pt_rebin
    )

    return source


# ------------------------------------------------------------
# Project one pT spectrum from one phi/eta cell
# ------------------------------------------------------------

def project_cell_pt_spectrum(
    h3,
    phi_bin,
    eta_bin,
    name
):
    spectrum = h3.ProjectionZ(
        name,
        phi_bin,
        phi_bin,
        eta_bin,
        eta_bin,
        "e"
    )

    spectrum.SetDirectory(0)
    spectrum.Sumw2()

    return spectrum

```

```python
def make_pt_slope_map(
    h3,
    name,
    fit_config
):
    phi_axis = h3.GetXaxis()
    eta_axis = h3.GetYaxis()

    slope_map = root.TH2D(
        name,
        "",
        phi_axis.GetNbins(),
        phi_axis.GetXmin(),
        phi_axis.GetXmax(),
        eta_axis.GetNbins(),
        eta_axis.GetXmin(),
        eta_axis.GetXmax()
    )

    slope_map.SetDirectory(0)
    slope_map.Sumw2()

    keep_spectra = []
    keep_graphs = []
    keep_fits = []

    for iphi in range(1, phi_axis.GetNbins() + 1):
        for ieta in range(1, eta_axis.GetNbins() + 1):

            spectrum = project_cell_pt_spectrum(
                h3,
                iphi,
                ieta,
                f"{name}_phi{iphi}_eta{ieta}_spectrum"
            )

            slope, slope_error, graph, fit_function = (
                fit_pt_spectrum_slope(
                    spectrum,
                    f"{name}_phi{iphi}_eta{ieta}",
                    fit_config
                )
            )

            keep_spectra.append(spectrum)

            if graph is not None:
                keep_graphs.append(graph)

            if fit_function is not None:
                keep_fits.append(fit_function)

            if not math.isfinite(slope):
                continue

            slope_map.SetBinContent(iphi, ieta, slope)
            slope_map.SetBinError(iphi, ieta, slope_error)

    slope_map._spectra = keep_spectra
    slope_map._graphs = keep_graphs
    slope_map._fits = keep_fits

    return slope_map
```

```python


# ------------------------------------------------------------
# Safe slope ratios
# ------------------------------------------------------------

def make_slope_ratio_2d(
    plus_map,
    minus_map,
    name
):
    ratio = plus_map.Clone(name)
    ratio.SetDirectory(0)
    ratio.Reset("ICES")
    ratio.Sumw2()

    for ix in range(1, ratio.GetNbinsX() + 1):
        for iy in range(1, ratio.GetNbinsY() + 1):

            plus = plus_map.GetBinContent(ix, iy)
            minus = minus_map.GetBinContent(ix, iy)

            plus_error = plus_map.GetBinError(ix, iy)
            minus_error = minus_map.GetBinError(ix, iy)

            if plus == 0.0 or minus == 0.0:
                continue

            if not math.isfinite(plus) or not math.isfinite(minus):
                continue

            value = plus / minus

            relative_error_squared = (
                (plus_error / plus) ** 2
                + (minus_error / minus) ** 2
            )

            ratio.SetBinContent(ix, iy, value)
            ratio.SetBinError(
                ix,
                iy,
                abs(value)
                * math.sqrt(relative_error_squared)
            )

    ratio._plus_map = plus_map
    ratio._minus_map = minus_map

    return ratio


def make_slope_ratio_1d(
    plus_hist,
    minus_hist,
    name
):
    ratio = plus_hist.Clone(name)
    ratio.SetDirectory(0)
    ratio.Reset("ICES")
    ratio.Sumw2()

    for ibin in range(1, ratio.GetNbinsX() + 1):

        plus = plus_hist.GetBinContent(ibin)
        minus = minus_hist.GetBinContent(ibin)

        plus_error = plus_hist.GetBinError(ibin)
        minus_error = minus_hist.GetBinError(ibin)

        if plus == 0.0 or minus == 0.0:
            continue

        if not math.isfinite(plus) or not math.isfinite(minus):
            continue

        value = plus / minus

        relative_error_squared = (
            (plus_error / plus) ** 2
            + (minus_error / minus) ** 2
        )

        ratio.SetBinContent(ibin, value)
        ratio.SetBinError(
            ibin,
            abs(value)
            * math.sqrt(relative_error_squared)
        )

    ratio._plus_hist = plus_hist
    ratio._minus_hist = minus_hist

    return ratio


# ------------------------------------------------------------
# Slope versus phi, integrated over eta
# ------------------------------------------------------------

def make_slope_vs_phi(
    h3,
    name,
    fit_config
):
    phi_axis = h3.GetXaxis()

    result = root.TH1D(
        name,
        "",
        phi_axis.GetNbins(),
        phi_axis.GetXmin(),
        phi_axis.GetXmax()
    )

    result.SetDirectory(0)
    result.Sumw2()

    keep_spectra = []
    keep_graphs = []
    keep_fits = []

    for iphi in range(1, phi_axis.GetNbins() + 1):

        spectrum = h3.ProjectionZ(
            f"{name}_phi{iphi}_spectrum",
            iphi,
            iphi,
            1,
            h3.GetNbinsY(),
            "e"
        )

        spectrum.SetDirectory(0)
        spectrum.Sumw2()

        slope, slope_error, graph, fit_function = (
            fit_pt_spectrum_slope(
                spectrum,
                f"{name}_phi{iphi}",
                fit_config
            )
        )

        keep_spectra.append(spectrum)

        if graph is not None:
            keep_graphs.append(graph)

        if fit_function is not None:
            keep_fits.append(fit_function)

        if not math.isfinite(slope):
            continue

        result.SetBinContent(iphi, slope)
        result.SetBinError(iphi, slope_error)

    result._spectra = keep_spectra
    result._graphs = keep_graphs
    result._fits = keep_fits

    return result


def make_slope_vs_eta(
    h3,
    name,
    fit_config
):
    eta_axis = h3.GetYaxis()

    result = root.TH1D(
        name,
        "",
        eta_axis.GetNbins(),
        eta_axis.GetXmin(),
        eta_axis.GetXmax()
    )

    result.SetDirectory(0)
    result.Sumw2()

    keep_spectra = []
    keep_graphs = []
    keep_fits = []

    for ieta in range(1, eta_axis.GetNbins() + 1):

        spectrum = h3.ProjectionZ(
            f"{name}_eta{ieta}_spectrum",
            1,
            h3.GetNbinsX(),
            ieta,
            ieta,
            "e"
        )

        spectrum.SetDirectory(0)
        spectrum.Sumw2()

        slope, slope_error, graph, fit_function = (
            fit_pt_spectrum_slope(
                spectrum,
                f"{name}_eta{ieta}",
                fit_config
            )
        )

        keep_spectra.append(spectrum)

        if graph is not None:
            keep_graphs.append(graph)

        if fit_function is not None:
            keep_fits.append(fit_function)

        if not math.isfinite(slope):
            continue

        result.SetBinContent(ieta, slope)
        result.SetBinError(ieta, slope_error)

    result._spectra = keep_spectra
    result._graphs = keep_graphs
    result._fits = keep_fits

    return result

```

```python
# ============================================================
# Produce slope-ratio QA for all configured pT fit intervals
# ============================================================

pt_slope_ratio_maps = {}
pt_slope_ratio_phi = {}
pt_slope_ratio_eta = {}

pt_slope_map_canvases = []
pt_slope_projection_canvases = []

for fit_config in spectrum_fit_ranges:

    range_name = fit_config["name"]
    range_label = fit_config["label"]

    # ========================================================
    # 2D phi-versus-eta slope-ratio maps
    # ========================================================

    canvas_map = root.TCanvas(
        f"c_pt_slope_ratio_phi_eta_{range_name}",
        f"c_pt_slope_ratio_phi_eta_{range_name}",
        1500,
        650
    )

    canvas_map.Divide(2, 1)

    keep_map_objects = []

    for ipad, (plus_dir, minus_dir, side_name) in enumerate(side_pairs):

        hplus3 = prepare_slope_source(
            hists_dict[plus_dir][pt_phi_eta_hist_name],
            f"hplus3_slope_{side_name}_{range_name}",
            fit_config
        )

        hminus3 = prepare_slope_source(
            hists_dict[minus_dir][pt_phi_eta_hist_name],
            f"hminus3_slope_{side_name}_{range_name}",
            fit_config
        )

        slope_plus_map = make_pt_slope_map(
            hplus3,
            f"h_pt_slope_plus_map_{side_name}_{range_name}",
            fit_config
        )

        slope_minus_map = make_pt_slope_map(
            hminus3,
            f"h_pt_slope_minus_map_{side_name}_{range_name}",
            fit_config
        )

        ratio_map = make_slope_ratio_2d(
            slope_plus_map,
            slope_minus_map,
            f"h_pt_slope_ratio_map_{side_name}_{range_name}"
        )

        pt_slope_ratio_maps[
            (range_name, side_name)
        ] = ratio_map

        # Integrated slope versus phi.
        slope_plus_phi = make_slope_vs_phi(
            hplus3,
            f"h_pt_slope_plus_phi_{side_name}_{range_name}",
            fit_config
        )

        slope_minus_phi = make_slope_vs_phi(
            hminus3,
            f"h_pt_slope_minus_phi_{side_name}_{range_name}",
            fit_config
        )

        ratio_phi = make_slope_ratio_1d(
            slope_plus_phi,
            slope_minus_phi,
            f"h_pt_slope_ratio_phi_{side_name}_{range_name}"
        )

        # Integrated slope versus eta.
        slope_plus_eta = make_slope_vs_eta(
            hplus3,
            f"h_pt_slope_plus_eta_{side_name}_{range_name}",
            fit_config
        )

        slope_minus_eta = make_slope_vs_eta(
            hminus3,
            f"h_pt_slope_minus_eta_{side_name}_{range_name}",
            fit_config
        )

        ratio_eta = make_slope_ratio_1d(
            slope_plus_eta,
            slope_minus_eta,
            f"h_pt_slope_ratio_eta_{side_name}_{range_name}"
        )

        pt_slope_ratio_phi[
            (range_name, side_name)
        ] = ratio_phi

        pt_slope_ratio_eta[
            (range_name, side_name)
        ] = ratio_eta

        canvas_map.cd(ipad + 1)

        root.gPad.SetLeftMargin(0.12)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetRightMargin(0.17)
        root.gPad.SetTopMargin(0.11)

        Format_hist_loc(
            ratio_map,
            title=(
                f"{side_name}: p_{{T}} slope positive/negative, "
                f"{range_label}"
            ),
            x_title="#phi = atan2(p_{x},p_{y}) [rad]",
            y_title="#eta",
            x_title_offset=0.95,
            y_title_offset=0.95,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.052,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_map.GetYaxis().SetRangeUser(*eta_draw_range)

        ratio_map.GetZaxis().SetTitle(
            "slope positive / slope negative"
        )
        ratio_map.GetZaxis().CenterTitle()
        ratio_map.GetZaxis().SetTitleOffset(1.25)
        ratio_map.GetZaxis().SetTitleSize(0.043)
        ratio_map.GetZaxis().SetLabelSize(0.04)

        ratio_map.SetMinimum(slope_ratio_2d_range[0])
        ratio_map.SetMaximum(slope_ratio_2d_range[1])

        ratio_map.Draw("COLZ")

        keep_map_objects += [
            hplus3,
            hminus3,
            slope_plus_map,
            slope_minus_map,
            ratio_map,
            slope_plus_phi,
            slope_minus_phi,
            ratio_phi,
            slope_plus_eta,
            slope_minus_eta,
            ratio_eta,
        ]

    canvas_map.Update()
    canvas_map.Draw()

    canvas_map._objects = keep_map_objects
    pt_slope_map_canvases.append(canvas_map)

    if safe_to_pdf:
        canvas_map.Print(
            outdir
            + f"/qa_pt_slope_ratio_phi_eta_{range_name}.pdf"
        )

    # ========================================================
    # Slope-ratio projections versus phi and eta
    # ========================================================

    canvas_projection = root.TCanvas(
        f"c_pt_slope_ratio_projections_{range_name}",
        f"c_pt_slope_ratio_projections_{range_name}",
        1500,
        1100
    )

    canvas_projection.Divide(2, 2)

    keep_projection_objects = []
    keep_lines = []

    for iside, (_, _, side_name) in enumerate(side_pairs):

        ratio_phi = pt_slope_ratio_phi[
            (range_name, side_name)
        ]

        ratio_eta = pt_slope_ratio_eta[
            (range_name, side_name)
        ]

        for ratio in [ratio_phi, ratio_eta]:
            ratio.SetLineColor(root.kBlack)
            ratio.SetMarkerColor(root.kBlack)
            ratio.SetMarkerStyle(20)
            ratio.SetMarkerSize(0.9)
            ratio.SetLineWidth(2)

        # Phi projection
        canvas_projection.cd(2 * iside + 1)

        root.gPad.SetLeftMargin(0.13)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetGridy()

        Format_hist_loc(
            ratio_phi,
            title=(
                f"{side_name}: slope positive/negative, "
                f"{range_label}"
            ),
            x_title="#phi = atan2(p_{x},p_{y}) [rad]",
            y_title="slope positive / slope negative",
            x_title_offset=0.95,
            y_title_offset=1.08,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.048,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_phi.SetMinimum(slope_ratio_1d_range[0])
        ratio_phi.SetMaximum(slope_ratio_1d_range[1])
        ratio_phi.Draw("E1")

        phi_line = root.TLine(
            -root.TMath.Pi(),
            1.0,
            root.TMath.Pi(),
            1.0
        )

        phi_line.SetLineStyle(2)
        phi_line.SetLineWidth(2)
        phi_line.SetLineColor(root.kGray + 2)
        phi_line.Draw("same")

        # Eta projection
        canvas_projection.cd(2 * iside + 2)

        root.gPad.SetLeftMargin(0.13)
        root.gPad.SetBottomMargin(0.13)
        root.gPad.SetGridy()

        Format_hist_loc(
            ratio_eta,
            title=(
                f"{side_name}: slope positive/negative, "
                f"{range_label}"
            ),
            x_title="#eta",
            y_title="slope positive / slope negative",
            x_title_offset=0.95,
            y_title_offset=1.08,
            center_title=True,
            x_title_size=0.052,
            y_title_size=0.048,
            label_size_x=0.043,
            label_size_y=0.043
        )

        ratio_eta.GetXaxis().SetRangeUser(*eta_draw_range)
        ratio_eta.SetMinimum(slope_ratio_1d_range[0])
        ratio_eta.SetMaximum(slope_ratio_1d_range[1])
        ratio_eta.Draw("E1")

        eta_line = root.TLine(
            eta_draw_range[0],
            1.0,
            eta_draw_range[1],
            1.0
        )

        eta_line.SetLineStyle(2)
        eta_line.SetLineWidth(2)
        eta_line.SetLineColor(root.kGray + 2)
        eta_line.Draw("same")

        keep_projection_objects += [
            ratio_phi,
            ratio_eta
        ]

        keep_lines += [
            phi_line,
            eta_line
        ]

    canvas_projection.Update()
    canvas_projection.Draw()

    canvas_projection._objects = keep_projection_objects
    canvas_projection._lines = keep_lines

    pt_slope_projection_canvases.append(
        canvas_projection
    )

    if safe_to_pdf:
        canvas_projection.Print(
            outdir
            + f"/qa_pt_slope_ratio_projections_{range_name}.pdf"
        )
```
