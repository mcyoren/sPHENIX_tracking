```python
input_file = root.TFile(file_path+file_name)
hists = []
for idir in Directories:
    h_folder = []
    #input_file.cd(idir)
    for hist_name in hist_names:
        h_folder.append(input_file.Get(idir+"/"+hist_name))
        h_folder[-1].SetDirectory(0)
        h_folder[-1].SetName(hist_name+"_"+idir)
    hists.append(h_folder)
input_file.Close()
```

```python
#going form hist array to hist dictionary
hists_dict = {}
for i, idir in enumerate(Directories):
    hists_dict[idir] = {}
    for j, hist_name in enumerate(hist_names):
        hists_dict[idir][hist_name] = hists[i][j]
```

```python
def Format_hist_loc(h, title, x_title, y_title, x_title_size=0.06, y_title_size=0.06, x_title_offset=1.0, \
                    y_title_offset=1.0, label_size_x=0.06, label_size_y=0.06, label_size_z=0.06,\
                         z_axis_offset=1.0, z_range = None, center_title=False, title_size=0.06):
    if h is None:
        print("Error: Histogram is None. Cannot format.")
        return
    if title is not None:
        h.SetTitle(title)
    if x_title is not None:
        h.GetXaxis().SetTitle(x_title)
    if y_title is not None:
        h.GetYaxis().SetTitle(y_title)
    h.GetXaxis().SetTitleSize(x_title_size)
    h.GetYaxis().SetTitleSize(y_title_size)
    h.GetXaxis().SetTitleOffset(x_title_offset)
    h.GetYaxis().SetTitleOffset(y_title_offset)
    h.GetXaxis().SetLabelSize(label_size_x)
    h.GetYaxis().SetLabelSize(label_size_y)
    h.GetZaxis().SetLabelSize(label_size_z)
    h.GetZaxis().SetTitleOffset(z_axis_offset)
    #h.SetTitleSize(title_size)
    if z_range is not None:
        h.GetZaxis().SetRangeUser(z_range[0], z_range[1])
    if center_title:
        h.GetXaxis().CenterTitle()
        h.GetYaxis().CenterTitle()
```

```python
#make rDCAzero vs pT projections from h_rDCA_vs_qOverPt_vs_pt
#
#original TH3 axes:
#X = q/pT
#Y = rDCAzero
#Z = pT
#
#pT binning:
#0.2 GeV/c bins from 0.2 to 3.0 GeV/c
#0.5 GeV/c bins from 3.0 to 5.0 GeV/c
#
#profile option "s" stores the standard deviation as the profile-bin error

from array import array
import math

hist_name = "h_rDCA_vs_qOverPt_vs_pt"

dirs_dca_pt = [
    'side0_qplus',
    'side0_qminus',
    'side1_qplus',
    'side1_qminus'
]

names_dca_pt = [
    'side0 positive',
    'side0 negative',
    'side1 positive',
    'side1 negative'
]

colors_dca_pt = [
    root.kRed+1,
    root.kBlue+1,
    root.kMagenta+1,
    root.kGreen+2
]

markers_dca_pt = [20, 21, 24, 25]

pt_edges = []

pt_value = 0.2
while pt_value < 3.0-1.e-8:
    pt_edges.append(pt_value)
    pt_value += 0.2

pt_edges.append(3.0)
pt_edges += [3.5, 4.0, 4.5, 5.0]

pt_edges_array = array('d', pt_edges)

dca_pt_2d = []
dca_pt_profiles = []
dca_pt_means = []
dca_pt_sigmas = []

for icase, idir in enumerate(dirs_dca_pt):

    h3 = hists_dict[idir][hist_name]

    #keep the original rDCAzero binning, restricted to -9 < rDCAzero < 9 cm
    first_dca_bin = h3.GetYaxis().FindBin(-9. + 1.e-6)
    last_dca_bin  = h3.GetYaxis().FindBin( 9. - 1.e-6)

    n_dca_bins = last_dca_bin-first_dca_bin+1

    dca_low = h3.GetYaxis().GetBinLowEdge(first_dca_bin)
    dca_high = h3.GetYaxis().GetBinUpEdge(last_dca_bin)

    h2 = root.TH2D(
        f"h_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array,
        n_dca_bins,
        dca_low,
        dca_high
    )

    h2.SetDirectory(0)
    h2.Sumw2()

    #sum over all q/pT bins and rebin pT into the requested variable bins
    for iz in range(1, h3.GetNbinsZ()+1):

        pt = h3.GetZaxis().GetBinCenter(iz)

        if pt < pt_edges[0] or pt >= pt_edges[-1]:
            continue

        new_pt_bin = h2.GetXaxis().FindBin(pt)

        for iy in range(first_dca_bin, last_dca_bin+1):

            rdca = h3.GetYaxis().GetBinCenter(iy)

            content_sum = 0.
            error2_sum = 0.

            for ix in range(1, h3.GetNbinsX()+1):

                content = h3.GetBinContent(ix, iy, iz)
                error = h3.GetBinError(ix, iy, iz)

                content_sum += content
                error2_sum += error*error

            new_dca_bin = h2.GetYaxis().FindBin(rdca)

            old_content = h2.GetBinContent(new_pt_bin, new_dca_bin)
            old_error = h2.GetBinError(new_pt_bin, new_dca_bin)

            h2.SetBinContent(
                new_pt_bin,
                new_dca_bin,
                old_content+content_sum
            )

            h2.SetBinError(
                new_pt_bin,
                new_dca_bin,
                math.sqrt(old_error*old_error+error2_sum)
            )

    #profile mean rDCAzero versus pT
    #option "s": profile errors are standard deviations
    profile = h2.ProfileX(
        f"p_rDCAzero_vs_pt_{idir}",
        1,
        h2.GetNbinsY(),
        "s"
    )

    profile.SetDirectory(0)

    #separate mean histogram
    #mean uncertainty = sigma/sqrt(N)
    hmean = root.TH1D(
        f"h_mean_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array
    )

    hmean.SetDirectory(0)

    #separate sigma histogram
    hsigma = root.TH1D(
        f"h_sigma_rDCAzero_vs_pt_{idir}",
        "",
        len(pt_edges)-1,
        pt_edges_array
    )

    hsigma.SetDirectory(0)

    for ibin in range(1, profile.GetNbinsX()+1):

        mean = profile.GetBinContent(ibin)
        sigma = profile.GetBinError(ibin)
        entries = profile.GetBinEntries(ibin)

        mean_error = sigma/math.sqrt(entries) if entries > 0 else 0.

        hmean.SetBinContent(ibin, mean)
        hmean.SetBinError(ibin, mean_error)

        hsigma.SetBinContent(ibin, sigma)
        hsigma.SetBinError(ibin, 0.)

    h2.SetLineColor(colors_dca_pt[icase])
    profile.SetLineColor(colors_dca_pt[icase])
    profile.SetMarkerColor(colors_dca_pt[icase])
    profile.SetMarkerStyle(markers_dca_pt[icase])

    hmean.SetLineColor(colors_dca_pt[icase])
    hmean.SetLineWidth(2)
    hmean.SetMarkerColor(colors_dca_pt[icase])
    hmean.SetMarkerStyle(markers_dca_pt[icase])
    hmean.SetMarkerSize(1.0)

    hsigma.SetLineColor(colors_dca_pt[icase])
    hsigma.SetLineWidth(2)
    hsigma.SetMarkerColor(colors_dca_pt[icase])
    hsigma.SetMarkerStyle(markers_dca_pt[icase])
    hsigma.SetMarkerSize(1.0)

    dca_pt_2d.append(h2)
    dca_pt_profiles.append(profile)
    dca_pt_means.append(hmean)
    dca_pt_sigmas.append(hsigma)
```
