```python
def Format_hist_loc(h, title, x_title, y_title, x_title_size=0.06, y_title_size=0.06, x_title_offset=1.0, \
                    y_title_offset=1.0, label_size_x=0.06, label_size_y=0.06, label_size_z=0.06,\
                         z_axis_offset=1.0, z_range = None, center_title=False, title_size=0.06):
    if h is None:
        print("Error: Histogram is None. Cannot format.")
        return
    if title is not None:
        h.SetTitle(title)
    if x_title is not None:
        h.GetXaxis().SetTitle(x_title)
    if y_title is not None:
        h.GetYaxis().SetTitle(y_title)
    h.GetXaxis().SetTitleSize(x_title_size)
    h.GetYaxis().SetTitleSize(y_title_size)
    h.GetXaxis().SetTitleOffset(x_title_offset)
    h.GetYaxis().SetTitleOffset(y_title_offset)
    h.GetXaxis().SetLabelSize(label_size_x)
    h.GetYaxis().SetLabelSize(label_size_y)
    h.GetZaxis().SetLabelSize(label_size_z)
    h.GetZaxis().SetTitleOffset(z_axis_offset)
    #h.SetTitleSize(title_size)
    if z_range is not None:
        h.GetZaxis().SetRangeUser(z_range[0], z_range[1])
    if center_title:
        h.GetXaxis().CenterTitle()
        h.GetYaxis().CenterTitle()
```

```python
def make_geometric_like_background(h_pp, h_mm, name):

    if h_pp.GetNbinsX() != h_mm.GetNbinsX():
        raise RuntimeError("++ and -- histograms have different binning")

    h_bg = h_pp.Clone(name)
    h_bg.Reset("ICES")
    h_bg.SetDirectory(0)
    keep(h_bg)

    for ibin in range(1, h_bg.GetNbinsX()+1):

        npp = h_pp.GetBinContent(ibin)
        nmm = h_mm.GetBinContent(ibin)

        epp = h_pp.GetBinError(ibin)
        emm = h_mm.GetBinError(ibin)

        if npp > 0. and nmm > 0.:

            value = 2.*math.sqrt(npp*nmm)

            #error propagation for B = 2*sqrt(N++*N--)
            error2 = (
                (nmm/npp)*epp*epp +
                (npp/nmm)*emm*emm
            )

            error = math.sqrt(error2)

        else:
            value = 0.
            error = 0.

        h_bg.SetBinContent(ibin, value)
        h_bg.SetBinError(ibin, error)

    return h_bg
```

```python


def get_hist(path):
    h = input_file.Get(path)
    if not h:
        raise RuntimeError("Missing histogram: "+path)
    h = h.Clone(path.replace("/", "_")+"_clone")
    h.SetDirectory(0)
    return keep(h)

def hist_path(cut_name, charge_name, hist_name):
    return f"{cut_name}/{charge_name}/{hist_name}"

def sideband_integral(h, sidebands):
    total = 0.
    error2 = 0.

    for xmin, xmax in sidebands:
        b1 = h.GetXaxis().FindBin(xmin + 1.e-8)
        b2 = h.GetXaxis().FindBin(xmax - 1.e-8)

        for ibin in range(b1, b2 + 1):
            total += h.GetBinContent(ibin)
            error2 += h.GetBinError(ibin)**2

    return total, math.sqrt(error2)

def project_mass(h2, ptmin, ptmax, name):
    b1 = h2.GetXaxis().FindBin(ptmin+1.e-8)
    b2 = h2.GetXaxis().FindBin(ptmax-1.e-8)
    h = h2.ProjectionY(name, b1, b2, "e")
    h.SetDirectory(0)
    return keep(h)

def scale_like_to_unlike(h_unlike, h_like, sidebands):
    fg, fg_err = sideband_integral(h_unlike, sidebands)
    bg, bg_err = sideband_integral(h_like, sidebands)

    scale = fg/bg if bg > 0. else 1.
    scale_err = 0.

    if fg > 0. and bg > 0.:
        scale_err = scale*math.sqrt((fg_err/fg)**2+(bg_err/bg)**2)

    h_scaled = h_like.Clone(h_like.GetName()+"_scaled")
    h_scaled.SetDirectory(0)
    h_scaled.Scale(scale)
    keep(h_scaled)

    return h_scaled, scale, scale_err

def make_subtracted(h_unlike, h_like_scaled, name):
    h_sub = h_unlike.Clone(name)
    h_sub.SetDirectory(0)
    h_sub.Add(h_like_scaled, -1.)
    return keep(h_sub)

def setup_mass_hist(h, xtitle, ytitle="pairs"):
    Format_hist_loc(
        h,
        title="",
        x_title=xtitle,
        y_title=ytitle,
        x_title_offset=.9,
        y_title_offset=.9,
        center_title=True,
        x_title_size=0.055,
        y_title_size=0.055,
        label_size_x=0.045,
        label_size_y=0.045
    )

def draw_header(ptmin, ptmax, cut_name, extra=None, extra1 =None, x=0.16, size=0.06):
    latex = keep(root.TLatex())
    latex.SetNDC()
    latex.SetTextFont(42)

    latex.SetTextSize(size)
    latex.DrawLatex(x, 0.925, data_set+", "+run_text)

    latex.SetTextSize(size)
    latex.DrawLatex(x, 0.850, f"{ptmin:.1f} < p_{{T}}^{{V0}} < {ptmax:.1f} GeV/c")
    #latex.DrawLatex(x, 0.825, cut_name)

    if extra:
        latex.DrawLatex(x, 0.775, extra)

    if extra1:
        latex.DrawLatex(x, 0.700, extra1)

    return latex

def draw_zero_line(xmin, xmax):
    line = keep(root.TLine(xmin, 0., xmax, 0.))
    line.SetLineColor(root.kBlack)
    line.SetLineStyle(2)
    line.SetLineWidth(1)
    line.Draw("same")
    return line

def get_lambda_h2(cut_name, charge_name):
    if lambda_species == "Lambda":
        return get_hist(hist_path(cut_name, charge_name, "h_mass_Lambda_vs_v0_pt"))

    if lambda_species == "AntiLambda":
        return get_hist(hist_path(cut_name, charge_name, "h_mass_AntiLambda_vs_v0_pt"))

    h = get_hist(hist_path(cut_name, charge_name, "h_mass_Lambda_vs_v0_pt"))
    hbar = get_hist(hist_path(cut_name, charge_name, "h_mass_AntiLambda_vs_v0_pt"))
    h.Add(hbar)
    h.SetName(h.GetName()+"_LambdaPlusAntiLambda")
    return h
```

```python
# ============================================================
# Helpers for all-cut yield extraction
# ============================================================

def get_species_h2(cut_name, charge_name, species):
    cfg = yield_fit_config[species]
    return get_hist(
        hist_path(cut_name, charge_name, cfg["hist_name"])
    )


def prepare_one_spectrum(cut_name, species, ptmin, ptmax, rebin, index):
    cfg = yield_fit_config[species]

    h2_unlike = get_species_h2(cut_name, "unlike", species)
    h_fg = project_mass(
        h2_unlike,
        ptmin,
        ptmax,
        f"h_{species}_{cut_name}_fg_pt{index}"
    )

    if like_background_method == "combined_like":
        h2_like = get_species_h2(cut_name, "like", species)
        h_like_raw = project_mass(
            h2_like,
            ptmin,
            ptmax,
            f"h_{species}_{cut_name}_like_pt{index}"
        )

    elif like_background_method == "geometric":
        h2_pp = get_species_h2(cut_name, "plusplus", species)
        h2_mm = get_species_h2(cut_name, "minusminus", species)

        h_pp = project_mass(
            h2_pp,
            ptmin,
            ptmax,
            f"h_{species}_{cut_name}_pp_pt{index}"
        )
        h_mm = project_mass(
            h2_mm,
            ptmin,
            ptmax,
            f"h_{species}_{cut_name}_mm_pt{index}"
        )

    else:
        raise RuntimeError(
            "like_background_method must be 'combined_like' or 'geometric'"
        )

    if rebin > 1:
        h_fg.Rebin(rebin)

        if like_background_method == "combined_like":
            h_like_raw.Rebin(rebin)
        else:
            h_pp.Rebin(rebin)
            h_mm.Rebin(rebin)

    if like_background_method == "geometric":
        h_like_raw = make_geometric_like_background(
            h_pp,
            h_mm,
            f"h_{species}_{cut_name}_geometric_pt{index}"
        )

    h_bg, scale, scale_err = scale_like_to_unlike(
        h_fg,
        h_like_raw,
        cfg["sidebands"]
    )

    h_sub = make_subtracted(
        h_fg,
        h_bg,
        f"h_{species}_{cut_name}_sub_pt{index}"
    )

    return {
        "fg": h_fg,
        "like_raw": h_like_raw,
        "bg": h_bg,
        "sub": h_sub,
        "scale": scale,
        "scale_err": scale_err,
    }


def matrix_element(matrix, i, j):
    try:
        return float(matrix[i][j])
    except Exception:
        return float(matrix(i, j))


def fit_signal_yield(h, species, fit_name):
    cfg = yield_fit_config[species]

    gaus = keep(root.TF1(
        fit_name + "_seed",
        "gaus",
        cfg["gaus_range"][0],
        cfg["gaus_range"][1]
    ))

    peak_bin = h.GetMaximumBin()
    amplitude0 = max(h.GetBinContent(peak_bin), 1.0)
    mean0 = h.GetBinCenter(peak_bin)

    sigma0 = 0.010 if species == "k0s" else 0.006

    gaus.SetParameters(amplitude0, mean0, sigma0)
    gaus.SetParLimits(
        1,
        cfg["mean_limits"][0],
        cfg["mean_limits"][1]
    )
    gaus.SetParLimits(
        2,
        cfg["sigma_limits"][0],
        cfg["sigma_limits"][1]
    )

    h.Fit(gaus, "RQ0")

    full = keep(root.TF1(
        fit_name + "_full",
        "gaus(0)+pol2(3)",
        cfg["fit_range"][0],
        cfg["fit_range"][1]
    ))

    full.SetParameters(
        gaus.GetParameter(0),
        gaus.GetParameter(1),
        abs(gaus.GetParameter(2)),
        0.0,
        0.0,
        0.0
    )

    full.SetParLimits(
        1,
        cfg["mean_limits"][0],
        cfg["mean_limits"][1]
    )
    full.SetParLimits(
        2,
        cfg["sigma_limits"][0],
        cfg["sigma_limits"][1]
    )

    fit_ptr = h.Fit(full, "RQS0")
    fit_status = int(fit_ptr)

    amplitude = full.GetParameter(0)
    mean = full.GetParameter(1)
    sigma = abs(full.GetParameter(2))

    bin_width = h.GetXaxis().GetBinWidth(1)
    norm = math.sqrt(2.0 * math.pi) / bin_width
    yield_value = norm * amplitude * sigma

    # Error on A*sigma, including the A-sigma covariance.
    yield_error = float("nan")
    try:
        covariance = fit_ptr.GetCovarianceMatrix()
        var_a = matrix_element(covariance, 0, 0)
        var_sigma = matrix_element(covariance, 2, 2)
        cov_a_sigma = matrix_element(covariance, 0, 2)

        variance = norm**2 * (
            sigma**2 * var_a
            + amplitude**2 * var_sigma
            + 2.0 * amplitude * sigma * cov_a_sigma
        )
        yield_error = math.sqrt(max(variance, 0.0))
    except Exception:
        # Fallback when a covariance matrix is unavailable.
        amplitude_error = full.GetParError(0)
        sigma_error = full.GetParError(2)

        if amplitude != 0.0 and sigma != 0.0:
            yield_error = abs(yield_value) * math.sqrt(
                (amplitude_error / amplitude)**2
                + (sigma_error / sigma)**2
            )

    signal = keep(root.TF1(
        fit_name + "_signal",
        "gaus",
        cfg["fit_range"][0],
        cfg["fit_range"][1]
    ))
    signal.SetParameters(amplitude, mean, sigma)

    background = keep(root.TF1(
        fit_name + "_background",
        "pol2",
        cfg["fit_range"][0],
        cfg["fit_range"][1]
    ))
    background.SetParameters(
        full.GetParameter(3),
        full.GetParameter(4),
        full.GetParameter(5)
    )

    return {
        "yield": yield_value,
        "yield_error": yield_error,
        "mean": mean,
        "mean_error": full.GetParError(1),
        "sigma": sigma,
        "sigma_error": full.GetParError(2),
        "fit_status": fit_status,
        "full": full,
        "signal": signal,
        "background": background,
    }


def ratio_with_error(numerator, numerator_error, denominator, denominator_error):
    if denominator <= 0.0:
        return float("nan"), float("nan")

    ratio = numerator / denominator

    if numerator <= 0.0:
        return ratio, float("nan")

    error = abs(ratio) * math.sqrt(
        (numerator_error / numerator)**2
        + (denominator_error / denominator)**2
    )
    return ratio, error


def make_graph(name, title, rows, value_key, error_key):
    x = array("d", [row["pt_center"] for row in rows])
    ex = array("d", [row["pt_half_width"] for row in rows])
    y = array("d", [row[value_key] for row in rows])
    ey = array("d", [row[error_key] for row in rows])

    graph = keep(root.TGraphErrors(
        len(rows),
        x,
        y,
        ex,
        ey
    ))
    graph.SetName(name)
    graph.SetTitle(title)
    graph.SetMarkerStyle(20)
    graph.SetMarkerSize(1.0)
    graph.SetLineWidth(2)
    return graph
```

```python
# ============================================================
# Calculate K0S, Lambda, and anti-Lambda yields for every cut
# ============================================================

all_cut_results = {}
flat_rows = []
fit_objects_by_cut = {}

for cut_name in cuts:
    print("\nProcessing", cut_name)

    cut_rows = []
    fit_objects_by_cut[cut_name] = {}

    for ipt in range(len(pt_bins_yield) - 1):
        ptmin = pt_bins_yield[ipt]
        ptmax = pt_bins_yield[ipt + 1]

        species_results = {}

        for species in ["k0s", "Lambda", "AntiLambda"]:
            rebin = yield_fit_config[species]["rebin"][ipt]

            spectrum = prepare_one_spectrum(
                cut_name,
                species,
                ptmin,
                ptmax,
                rebin,
                ipt
            )

            fit_result = fit_signal_yield(
                spectrum["sub"],
                species,
                f"fit_{species}_{cut_name}_pt{ipt}"
            )

            species_results[species] = {
                **spectrum,
                **fit_result,
                "rebin": rebin,
            }

            fit_objects_by_cut[cut_name][(species, ipt)] = species_results[species]

        lambda_yield = species_results["Lambda"]["yield"]
        lambda_error = species_results["Lambda"]["yield_error"]

        antilambda_yield = species_results["AntiLambda"]["yield"]
        antilambda_error = species_results["AntiLambda"]["yield_error"]

        k0s_yield = species_results["k0s"]["yield"]
        k0s_error = species_results["k0s"]["yield_error"]

        anti_over_lambda, anti_over_lambda_error = ratio_with_error(
            antilambda_yield,
            antilambda_error,
            lambda_yield,
            lambda_error
        )

        if lambda_over_k0s_species == "Lambda":
            baryon_yield = lambda_yield
            baryon_error = lambda_error
        elif lambda_over_k0s_species == "LambdaPlusAntiLambda":
            baryon_yield = lambda_yield + antilambda_yield
            baryon_error = math.sqrt(
                lambda_error**2 + antilambda_error**2
            )
        else:
            raise RuntimeError(
                "lambda_over_k0s_species must be "
                "'Lambda' or 'LambdaPlusAntiLambda'"
            )

        lambda_over_k0s, lambda_over_k0s_error = ratio_with_error(
            baryon_yield,
            baryon_error,
            k0s_yield,
            k0s_error
        )

        row = {
            "cut": cut_name,
            "ptmin": ptmin,
            "ptmax": ptmax,
            "pt_center": 0.5 * (ptmin + ptmax),
            "pt_half_width": 0.5 * (ptmax - ptmin),

            "k0s_yield": k0s_yield,
            "k0s_yield_error": k0s_error,
            "k0s_mean": species_results["k0s"]["mean"],
            "k0s_sigma": species_results["k0s"]["sigma"],
            "k0s_fit_status": species_results["k0s"]["fit_status"],

            "lambda_yield": lambda_yield,
            "lambda_yield_error": lambda_error,
            "lambda_mean": species_results["Lambda"]["mean"],
            "lambda_sigma": species_results["Lambda"]["sigma"],
            "lambda_fit_status": species_results["Lambda"]["fit_status"],

            "antilambda_yield": antilambda_yield,
            "antilambda_yield_error": antilambda_error,
            "antilambda_mean": species_results["AntiLambda"]["mean"],
            "antilambda_sigma": species_results["AntiLambda"]["sigma"],
            "antilambda_fit_status": species_results["AntiLambda"]["fit_status"],

            "antilambda_over_lambda": anti_over_lambda,
            "antilambda_over_lambda_error": anti_over_lambda_error,

            "lambda_over_k0s": lambda_over_k0s,
            "lambda_over_k0s_error": lambda_over_k0s_error,
        }

        cut_rows.append(row)
        flat_rows.append(row)

        print(
            f"  {ptmin:.1f} < pT < {ptmax:.1f}: "
            f"K0S={k0s_yield:.1f} +/- {k0s_error:.1f}, "
            f"Lambda={lambda_yield:.1f} +/- {lambda_error:.1f}, "
            f"anti-Lambda={antilambda_yield:.1f} +/- {antilambda_error:.1f}, "
            f"anti-Lambda/Lambda={anti_over_lambda:.3f} +/- {anti_over_lambda_error:.3f}, "
            f"Lambda/K0S={lambda_over_k0s:.3f} +/- {lambda_over_k0s_error:.3f}"
        )

    all_cut_results[cut_name] = cut_rows
```

```python
# ============================================================
# Prepare detailed cut descriptions from cut_lines
# ============================================================

def get_cut_description_lines(cut_name):
    """
    Return cut_lines[cut_name] as a clean list of strings.

    Supported forms:
        cut_lines[cut_name] = ["p_{T}>0.2", "pair DCA < 1 cm"]
        cut_lines[cut_name] = "p_{T}>0.2; pair DCA < 1 cm"
    """

    if cut_name not in cut_lines:
        return [cut_name]

    description = cut_lines[cut_name]

    if isinstance(description, str):
        # Split a single string into separate selections
        description = description.replace("\n", ";")
        lines = [
            line.strip()
            for line in description.split(";")
            if line.strip()
        ]
    else:
        lines = [
            str(line).strip()
            for line in description
            if str(line).strip()
        ]

    return lines


# Only cuts that actually have graphs
valid_selected_cuts = [
    cut_name
    for cut_name in selected_cuts_to_plot
    if cut_name in comparison_graphs
]

cut_description_lines = {
    cut_name: get_cut_description_lines(cut_name)
    for cut_name in valid_selected_cuts
}


# ============================================================
# Find selections common to all selected cuts
# ============================================================

common_cut_lines = []

if valid_selected_cuts:

    first_cut_name = valid_selected_cuts[0]

    for line in cut_description_lines[first_cut_name]:

        if all(
            line in cut_description_lines[other_cut_name]
            for other_cut_name in valid_selected_cuts[1:]
        ):
            common_cut_lines.append(line)


# Remove common selections from the individual descriptions
cut_specific_lines = {}

for cut_name in valid_selected_cuts:

    cut_specific_lines[cut_name] = [
        line
        for line in cut_description_lines[cut_name]
        if line not in common_cut_lines
    ]


print("Common selections:")
for line in common_cut_lines:
    print("   ", line)

print("\nCut-dependent selections:")
for cut_name in valid_selected_cuts:
    print(cut_name)
    for line in cut_specific_lines[cut_name]:
        print("   ", line)

#adding pt>0.2 Gev to common_cut_lines
common_cut_lines.append("p_{T} > 0.2 GeV/c, Lproj>2.0 cm, Sailors cut: #phi^{+}-#phi^{-} > #phi_{cut}")
```

```python
# ============================================================
# First-approximation efficiency + branching-ratio correction
# ============================================================
#
# The table below is the K0S/Lambda reconstruction-efficiency ratio
# shown in the supplied "Tsallis-weighted ratios" table:
#
#     eff_ratio = epsilon(K0S) / epsilon(Lambda)
#
# Therefore, for a raw data ratio R_raw = N(Lambda)/N(K0S),
#
#     R_corr = R_raw
#              * [epsilon(K0S)/epsilon(Lambda)]
#              * [BR(K0S -> pi+pi-) / BR(Lambda -> p pi-)]
#
# The same factor is also applied to
# (Lambda + anti-Lambda)/(2 K0S), assuming the supplied Lambda
# efficiency ratio is applicable to anti-Lambda as a first approximation.
#
# IMPORTANT:
#   * The efficiency table exists only for cut02--cut07 and 0.5 < pT < 3.0.
#   * Existing raw-yield and raw-ratio results above are NOT changed.
#   * The current data pT bins are finer than the efficiency bins, so the
#     efficiency bin is selected from the DATA-BIN CENTER. This is explicitly
#     a first approximation; for the final result, use matched pT binning or
#     event/spectrum-weighted efficiencies.
#   * Branching-fraction uncertainties are not included below; change the
#     values here if you want to use a different PDG set.

br_k0s_to_pipi = 0.6920
br_lambda_to_ppi = 0.6390
br_ratio_k0s_over_lambda = br_k0s_to_pipi / br_lambda_to_ppi

# (ptmin, ptmax), with upper edge exclusive except for the final 3.0 edge
efficiency_ratio_pt_bins = [
    (0.5, 0.8),
    (0.8, 1.1),
    (1.1, 1.4),
    (1.4, 1.8),
    (1.8, 2.2),
    (2.2, 3.0),
]

# Values are (epsilon_K0S / epsilon_Lambda, uncertainty).
# None means the uncertainty was not readable from the supplied screenshot.
efficiency_ratio_k0s_over_lambda = {
    "cut02_preselection": [None,(6.009,.604),(3.230,.218),(1.811,.089),(1.350,.055),(1.243,.049)],
    "cut03_baseline": [None,(6.035,.622),(3.205,.219),(1.836,.092),(1.333,.055),(1.231,.049)],
    "cut04_pairDCA_15mm": [None,(6.200,.679),(3.238,.235),(1.850,.099),(1.368,.061),(1.224,.052)],
    "cut05_pairDCA_10mm": [None,(6.417,.768),(3.273,.258),(1.942,.114),(1.375,.066),(1.262,.058)],
    "cut06_pairDCA_7mm": [None,(6.032,.799),(3.320,.291),(2.018,.127),(1.429,.073),(1.274,.061)],
    "cut07_pairDCA_5mm": [None,(5.941,.849),(3.151,.302),(2.127,.146),(1.493,.081),(1.275,.065)],
}

def get_efficiency_ratio_for_pt(cut_name, pt_value):
    """Return (eff_ratio, eff_error, eff_ptmin, eff_ptmax) for one pT value."""
    cut_values = efficiency_ratio_k0s_over_lambda.get(cut_name)
    if cut_values is None:
        return None

    for ibin, ((ptmin, ptmax), entry) in enumerate(
        zip(efficiency_ratio_pt_bins, cut_values)
    ):
        in_bin = (ptmin <= pt_value < ptmax)
        if ibin == len(efficiency_ratio_pt_bins) - 1:
            in_bin = (ptmin <= pt_value <= ptmax)

        if in_bin:
            if entry is None:
                return None
            eff_ratio, eff_error = entry
            return eff_ratio, eff_error, ptmin, ptmax

    return None


def multiply_ratio_with_factor(
    raw_ratio,
    raw_error,
    eff_ratio,
    eff_error,
    branching_ratio_factor
):
    """
    Correct a raw Lambda/K0S-type ratio.

    Error includes raw-ratio and efficiency-ratio uncertainties.
    Branching-fraction uncertainty is intentionally omitted for this
    first approximation.
    """
    if not math.isfinite(raw_ratio) or not math.isfinite(eff_ratio):
        return float("nan"), float("nan")

    corrected = raw_ratio * eff_ratio * branching_ratio_factor

    # If the efficiency uncertainty is unavailable, keep the central value
    # and propagate only the raw statistical uncertainty. A flag stored in
    # each result row records that this uncertainty is incomplete.
    rel2 = 0.0
    have_error = False

    if (
        raw_error is not None
        and math.isfinite(raw_error)
        and raw_ratio != 0.0
    ):
        rel2 += (raw_error / raw_ratio)**2
        have_error = True

    if (
        eff_error is not None
        and math.isfinite(eff_error)
        and eff_ratio != 0.0
    ):
        rel2 += (eff_error / eff_ratio)**2
        have_error = True

    corrected_error = (
        abs(corrected) * math.sqrt(rel2)
        if have_error
        else float("nan")
    )

    return corrected, corrected_error


# Add corrected quantities to the existing rows without changing any raw values.
for cut_name, rows in all_cut_results.items():
    for row in rows:
        eff_info = get_efficiency_ratio_for_pt(
            cut_name,
            row["pt_center"]
        )

        # Defaults for cuts/pT ranges without an efficiency correction.
        row["eff_k0s_over_lambda"] = float("nan")
        row["eff_k0s_over_lambda_error"] = float("nan")
        row["eff_ptmin"] = float("nan")
        row["eff_ptmax"] = float("nan")
        row["efficiency_error_complete"] = False
        row["br_k0s_over_lambda"] = br_ratio_k0s_over_lambda
        row["lambda_over_k0s_corrected"] = float("nan")
        row["lambda_over_k0s_corrected_error"] = float("nan")
        row["average_lambda_over_k0s"] = float("nan")
        row["average_lambda_over_k0s_error"] = float("nan")
        row["average_lambda_over_k0s_corrected"] = float("nan")
        row["average_lambda_over_k0s_corrected_error"] = float("nan")

        # Raw average ratio already used later in the notebook:
        # (Lambda + anti-Lambda)/(2 K0S)
        numerator = row["lambda_yield"] + row["antilambda_yield"]
        numerator_error = math.sqrt(
            row["lambda_yield_error"]**2
            + row["antilambda_yield_error"]**2
        )
        denominator = 2.0 * row["k0s_yield"]
        denominator_error = 2.0 * row["k0s_yield_error"]

        average_raw, average_raw_error = ratio_with_error(
            numerator,
            numerator_error,
            denominator,
            denominator_error
        )

        row["average_lambda_over_k0s"] = average_raw
        row["average_lambda_over_k0s_error"] = average_raw_error

        if eff_info is None:
            continue

        eff_ratio, eff_error, eff_ptmin, eff_ptmax = eff_info

        row["eff_k0s_over_lambda"] = eff_ratio
        row["eff_k0s_over_lambda_error"] = (
            eff_error if eff_error is not None else float("nan")
        )
        row["eff_ptmin"] = eff_ptmin
        row["eff_ptmax"] = eff_ptmax
        row["efficiency_error_complete"] = (eff_error is not None)

        corrected, corrected_error = multiply_ratio_with_factor(
            row["lambda_over_k0s"],
            row["lambda_over_k0s_error"],
            eff_ratio,
            eff_error,
            br_ratio_k0s_over_lambda
        )

        average_corrected, average_corrected_error = multiply_ratio_with_factor(
            average_raw,
            average_raw_error,
            eff_ratio,
            eff_error,
            br_ratio_k0s_over_lambda
        )

        row["lambda_over_k0s_corrected"] = corrected
        row["lambda_over_k0s_corrected_error"] = corrected_error
        row["average_lambda_over_k0s_corrected"] = average_corrected
        row["average_lambda_over_k0s_corrected_error"] = average_corrected_error

print("Branching-ratio factor BR(K0S->pi pi)/BR(Lambda->p pi) =", br_ratio_k0s_over_lambda)
print("Added first-approximation efficiency+BR corrected ratios to all_cut_results.")
```

```python
# ============================================================
# Diagnostic: pT>0.3 sample / nominal sample
#
# Top:
#   Lambda raw yield ratio
#   K0S raw yield ratio
#
# Bottom:
#   Lambda efficiency ratio
#   K0S efficiency ratio
#
# Ratio convention everywhere:
#       pt03 / nominal
# ============================================================

from array import array
import math
import ROOT as root

cuts_to_draw = [
    #"cut02_preselection",
    "cut03_baseline",
    "cut04_pairDCA_15mm",
    "cut05_pairDCA_10mm",
    "cut06_pairDCA_7mm",
    "cut07_pairDCA_5mm",
]


def ratio_err(a, ea, b, eb):
    if not all(math.isfinite(v) for v in [a, ea, b, eb]) or a <= 0 or b <= 0: return float("nan"), float("nan")
    r = a / b
    return r, abs(r) * math.sqrt((ea/a)**2 + (eb/b)**2)


def rows_by_pt(rows):
    return {round(r["pt_center"], 6): r for r in rows}


def make_ratio_graph(x, y, ey, name):
    if not x:
        g = keep(root.TGraphErrors())
    else:
        zero = array('d', [0.0] * len(x))
        g = keep(root.TGraphErrors(len(x), array('d', x), array('d', y), zero, array('d', ey)))
    g.SetName(name)
    return g


# ============================================================
# Build diagnostic graphs
# ============================================================

diag = {}

for icut, cut in enumerate(cuts_to_draw):

    if cut not in sample_results["nominal"] or cut not in sample_results["pt03"]:
        print("Skipping missing cut:", cut)
        continue

    nom = rows_by_pt(sample_results["nominal"][cut])
    p03 = rows_by_pt(sample_results["pt03"][cut])
    common_pt = sorted(set(nom) & set(p03))

    x_l, y_l, ey_l = [], [], []
    x_k, y_k, ey_k = [], [], []
    x_le, y_le, ey_le = [], [], []
    x_ke, y_ke, ey_ke = [], [], []

    for pt in common_pt:

        rn, rp = nom[pt], p03[pt]

        # Lambda raw yield ratio
        r, e = ratio_err(rp["lambda_yield"], rp["lambda_yield_error"], rn["lambda_yield"], rn["lambda_yield_error"])
        if math.isfinite(r):
            x_l.append(pt); y_l.append(r); ey_l.append(e)

        # K0S raw yield ratio
        r, e = ratio_err(rp["k0s_yield"], rp["k0s_yield_error"], rn["k0s_yield"], rn["k0s_yield_error"])
        if math.isfinite(r):
            x_k.append(pt); y_k.append(r); ey_k.append(e)

        # Lambda efficiency ratio
        en = get_eff(eff_nominal, "lambda", cut, pt)
        ep = get_eff(eff_pt03, "lambda", cut, pt)

        if en is not None and ep is not None:
            eff_n, deff_n = en
            eff_p, deff_p = ep
            r, e = ratio_err(eff_p, deff_p, eff_n, deff_n)
            if math.isfinite(r):
                x_le.append(pt); y_le.append(r); ey_le.append(e)

        # K0S efficiency ratio
        en = get_eff(eff_nominal, "k0s", cut, pt)
        ep = get_eff(eff_pt03, "k0s", cut, pt)

        if en is not None and ep is not None:
            eff_n, deff_n = en
            eff_p, deff_p = ep
            r, e = ratio_err(eff_p, deff_p, eff_n, deff_n)
            if math.isfinite(r):
                x_ke.append(pt); y_ke.append(r); ey_ke.append(e)

    diag[cut] = {
        "lambda_raw": make_ratio_graph(x_l, y_l, ey_l, f"g_diag_lambda_raw_{cut}"),
        "k0s_raw": make_ratio_graph(x_k, y_k, ey_k, f"g_diag_k0s_raw_{cut}"),
        "lambda_eff": make_ratio_graph(x_le, y_le, ey_le, f"g_diag_lambda_eff_{cut}"),
        "k0s_eff": make_ratio_graph(x_ke, y_ke, ey_ke, f"g_diag_k0s_eff_{cut}"),
    }


# ============================================================
# Canvas
# ============================================================

c_pt03_diag = keep(root.TCanvas("c_pt03_vs_nominal_diagnostic", "pt03 vs nominal diagnostic", 1500, 1100))
c_pt03_diag.Divide(2, 2)


def draw_panel(pad, key, ytitle, ymin, ymax, draw_legend=False):

    c_pt03_diag.cd(pad)

    root.gPad.SetLeftMargin(0.14)
    root.gPad.SetBottomMargin(0.14)
    root.gPad.SetRightMargin(0.04)
    root.gPad.SetTopMargin(0.08)

    frame = keep(root.TH1F(f"frame_diag_{pad}", "", 100, 0.75, 3.05))
    frame.SetDirectory(0)
    frame.SetMinimum(ymin)
    frame.SetMaximum(ymax)

    Format_hist_loc(frame, "", "p_{T} [GeV/c]", ytitle, center_title=True, x_title_offset=1.0, y_title_offset=1.05)
    frame.Draw()

    line = keep(root.TLine(0.75, 1.0, 3.05, 1.0))
    line.SetLineStyle(2)
    line.SetLineWidth(2)
    line.Draw()

    legend = None

    if draw_legend:
        legend = keep(root.TLegend(0.58, 0.55, 0.91, 0.88))
        legend.SetBorderSize(0)
        legend.SetFillStyle(0)
        legend.SetTextFont(42)
        legend.SetTextSize(0.030)

    for icut, cut in enumerate(cuts_to_draw):

        if cut not in diag: continue

        g = diag[cut][key]
        color = comparison_colors[icut % len(comparison_colors)]
        marker = comparison_markers[icut % len(comparison_markers)]

        g.SetMarkerColor(color)
        g.SetLineColor(color)
        g.SetMarkerStyle(marker)
        g.SetMarkerSize(1.1)
        g.SetLineWidth(2)
        g.Draw("PZ SAME")

        if legend is not None: legend.AddEntry(g, cut, "lp")

    if legend is not None: legend.Draw()


# ============================================================
# Draw 4 panels
# ============================================================

draw_panel(1, "lambda_raw", "N_{#Lambda}^{p_{T}>0.3} / N_{#Lambda}^{nominal}", 0.0, 1.9, True)
draw_panel(2, "k0s_raw", "N_{K^{0}_{S}}^{p_{T}>0.3} / N_{K^{0}_{S}}^{nominal}", 0.0, 1.9)

draw_panel(3, "lambda_eff", "#epsilon_{#Lambda}^{p_{T}>0.3} / #epsilon_{#Lambda}^{nominal}", 0.0, 1.9)
draw_panel(4, "k0s_eff", "#epsilon_{K^{0}_{S}}^{p_{T}>0.3} / #epsilon_{K^{0}_{S}}^{nominal}", 0.0, 1.9)

c_pt03_diag.Update()
c_pt03_diag.Draw()
```

```python
# ============================================================
# Double ratio:
#
#   [N(pt03)/N(nominal)] / [eff(pt03)/eff(nominal)]
#
# = [N(pt03)/eff(pt03)] / [N(nominal)/eff(nominal)]
#
# Top:    Lambda
# Bottom: K0S
# ============================================================

from array import array
import math
import ROOT as root


def double_ratio_graph(cut, species, name):

    nom = rows_by_pt(sample_results["nominal"][cut])
    p03 = rows_by_pt(sample_results["pt03"][cut])
    common_pt = sorted(set(nom) & set(p03))

    x, y, ey = [], [], []

    for pt in common_pt:
        rn, rp = nom[pt], p03[pt]

        if species == "lambda":
            yn, dyn = rn["lambda_yield"], rn["lambda_yield_error"]
            yp, dyp = rp["lambda_yield"], rp["lambda_yield_error"]
        else:
            yn, dyn = rn["k0s_yield"], rn["k0s_yield_error"]
            yp, dyp = rp["k0s_yield"], rp["k0s_yield_error"]

        en = get_eff(eff_nominal, species, cut, pt)
        ep = get_eff(eff_pt03, species, cut, pt)

        if en is None or ep is None: continue

        eff_n, deff_n = en
        eff_p, deff_p = ep

        ry, ery = ratio_err(yp, dyp, yn, dyn)
        re, ere = ratio_err(eff_p, deff_p, eff_n, deff_n)

        if not math.isfinite(ry) or not math.isfinite(re) or re <= 0: continue

        dr = ry / re
        edr = abs(dr) * math.sqrt((ery/ry)**2 + (ere/re)**2)

        x.append(pt)
        y.append(dr)
        ey.append(edr)

    zero = array('d', [0.0] * len(x))
    g = keep(root.TGraphErrors(len(x), array('d', x), array('d', y), zero, array('d', ey))) if x else keep(root.TGraphErrors())
    g.SetName(name)
    return g


# ------------------------------------------------------------
# Build graphs
# ------------------------------------------------------------

double_ratio = {}

for cut in cuts_to_draw:
    if cut not in sample_results["nominal"] or cut not in sample_results["pt03"]: continue

    double_ratio[cut] = {
        "lambda": double_ratio_graph(cut, "lambda", f"g_double_ratio_lambda_{cut}"),
        "k0s": double_ratio_graph(cut, "k0s", f"g_double_ratio_k0s_{cut}")
    }


# ------------------------------------------------------------
# Canvas
# ------------------------------------------------------------

c_double_ratio = keep(root.TCanvas("c_pt03_double_ratio", "pt03 corrected-yield double ratio", 900, 1000))
c_double_ratio.Divide(1, 2)


def draw_double_ratio_panel(pad, species, ytitle, draw_legend=False):

    c_double_ratio.cd(pad)

    root.gPad.SetLeftMargin(0.14)
    root.gPad.SetBottomMargin(0.14)
    root.gPad.SetRightMargin(0.04)
    root.gPad.SetTopMargin(0.08)

    frame = keep(root.TH1F(f"frame_double_ratio_{species}", "", 100, 0.75, 3.05))
    frame.SetDirectory(0)
    frame.SetMinimum(0.0)
    frame.SetMaximum(2.0)

    Format_hist_loc(frame, "", "p_{T} [GeV/c]", ytitle, center_title=True, x_title_offset=1.0, y_title_offset=1.05)
    frame.Draw()

    line = keep(root.TLine(0.75, 1.0, 3.05, 1.0))
    line.SetLineStyle(2)
    line.SetLineWidth(2)
    line.Draw()

    legend = None

    if draw_legend:
        legend = keep(root.TLegend(0.58, 0.55, 0.91, 0.88))
        legend.SetBorderSize(0)
        legend.SetFillStyle(0)
        legend.SetTextFont(42)
        legend.SetTextSize(0.030)

    for icut, cut in enumerate(cuts_to_draw):
        if cut not in double_ratio: continue

        g = double_ratio[cut][species]
        color = comparison_colors[icut % len(comparison_colors)]
        marker = comparison_markers[icut % len(comparison_markers)]

        g.SetMarkerColor(color)
        g.SetLineColor(color)
        g.SetMarkerStyle(marker)
        g.SetMarkerSize(1.1)
        g.SetLineWidth(2)
        g.Draw("PZ SAME")

        if legend is not None: legend.AddEntry(g, cut, "lp")

    if legend is not None: legend.Draw()


draw_double_ratio_panel(
    1,
    "lambda",
    "#frac{N_{#Lambda}^{p_{T}>0.3}/N_{#Lambda}^{nominal}}{#epsilon_{#Lambda}^{p_{T}>0.3}/#epsilon_{#Lambda}^{nominal}}",
    True
)

draw_double_ratio_panel(
    2,
    "k0s",
    "#frac{N_{K^{0}_{S}}^{p_{T}>0.3}/N_{K^{0}_{S}}^{nominal}}{#epsilon_{K^{0}_{S}}^{p_{T}>0.3}/#epsilon_{K^{0}_{S}}^{nominal}}"
)

c_double_ratio.Update()
c_double_ratio.Draw()
```
