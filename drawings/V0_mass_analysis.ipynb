```python
def Format_hist_loc(h, title, x_title, y_title, x_title_size=0.06, y_title_size=0.06, x_title_offset=1.0, \
                    y_title_offset=1.0, label_size_x=0.06, label_size_y=0.06, label_size_z=0.06,\
                         z_axis_offset=1.0, z_range = None, center_title=False, title_size=0.06):
    if h is None:
        print("Error: Histogram is None. Cannot format.")
        return
    if title is not None:
        h.SetTitle(title)
    if x_title is not None:
        h.GetXaxis().SetTitle(x_title)
    if y_title is not None:
        h.GetYaxis().SetTitle(y_title)
    h.GetXaxis().SetTitleSize(x_title_size)
    h.GetYaxis().SetTitleSize(y_title_size)
    h.GetXaxis().SetTitleOffset(x_title_offset)
    h.GetYaxis().SetTitleOffset(y_title_offset)
    h.GetXaxis().SetLabelSize(label_size_x)
    h.GetYaxis().SetLabelSize(label_size_y)
    h.GetZaxis().SetLabelSize(label_size_z)
    h.GetZaxis().SetTitleOffset(z_axis_offset)
    #h.SetTitleSize(title_size)
    if z_range is not None:
        h.GetZaxis().SetRangeUser(z_range[0], z_range[1])
    if center_title:
        h.GetXaxis().CenterTitle()
        h.GetYaxis().CenterTitle()
```

```python

#load file and helper functions

input_file = root.TFile.Open(file_path+file_name)

if not input_file or input_file.IsZombie():
    raise RuntimeError("Cannot open "+file_path+file_name)

ROOT_KEEPALIVE = []

def keep(obj):
    ROOT_KEEPALIVE.append(obj)
    root.SetOwnership(obj, False)
    return obj

```

```python
def make_geometric_like_background(h_pp, h_mm, name):

    if h_pp.GetNbinsX() != h_mm.GetNbinsX():
        raise RuntimeError("++ and -- histograms have different binning")

    h_bg = h_pp.Clone(name)
    h_bg.Reset("ICES")
    h_bg.SetDirectory(0)
    keep(h_bg)

    for ibin in range(1, h_bg.GetNbinsX()+1):

        npp = h_pp.GetBinContent(ibin)
        nmm = h_mm.GetBinContent(ibin)

        epp = h_pp.GetBinError(ibin)
        emm = h_mm.GetBinError(ibin)

        if npp > 0. and nmm > 0.:

            value = 2.*math.sqrt(npp*nmm)

            #error propagation for B = 2*sqrt(N++*N--)
            error2 = (
                (nmm/npp)*epp*epp +
                (npp/nmm)*emm*emm
            )

            error = math.sqrt(error2)

        else:
            value = 0.
            error = 0.

        h_bg.SetBinContent(ibin, value)
        h_bg.SetBinError(ibin, error)

    return h_bg
```

```python


def get_hist(path):
    h = input_file.Get(path)
    if not h:
        raise RuntimeError("Missing histogram: "+path)
    h = h.Clone(path.replace("/", "_")+"_clone")
    h.SetDirectory(0)
    return keep(h)

def hist_path(cut_name, charge_name, hist_name):
    return f"{cut_name}/{charge_name}/{hist_name}"

def sideband_integral(h, sidebands):
    total = 0.
    error2 = 0.

    for xmin, xmax in sidebands:
        b1 = h.GetXaxis().FindBin(xmin + 1.e-8)
        b2 = h.GetXaxis().FindBin(xmax - 1.e-8)

        for ibin in range(b1, b2 + 1):
            total += h.GetBinContent(ibin)
            error2 += h.GetBinError(ibin)**2

    return total, math.sqrt(error2)

def project_mass(h2, ptmin, ptmax, name):
    b1 = h2.GetXaxis().FindBin(ptmin+1.e-8)
    b2 = h2.GetXaxis().FindBin(ptmax-1.e-8)
    h = h2.ProjectionY(name, b1, b2, "e")
    h.SetDirectory(0)
    return keep(h)

def scale_like_to_unlike(h_unlike, h_like, sidebands):
    fg, fg_err = sideband_integral(h_unlike, sidebands)
    bg, bg_err = sideband_integral(h_like, sidebands)

    scale = fg/bg if bg > 0. else 1.
    scale_err = 0.

    if fg > 0. and bg > 0.:
        scale_err = scale*math.sqrt((fg_err/fg)**2+(bg_err/bg)**2)

    h_scaled = h_like.Clone(h_like.GetName()+"_scaled")
    h_scaled.SetDirectory(0)
    h_scaled.Scale(scale)
    keep(h_scaled)

    return h_scaled, scale, scale_err

def make_subtracted(h_unlike, h_like_scaled, name):
    h_sub = h_unlike.Clone(name)
    h_sub.SetDirectory(0)
    h_sub.Add(h_like_scaled, -1.)
    return keep(h_sub)

def setup_mass_hist(h, xtitle, ytitle="pairs"):
    Format_hist_loc(
        h,
        title="",
        x_title=xtitle,
        y_title=ytitle,
        x_title_offset=.9,
        y_title_offset=.9,
        center_title=True,
        x_title_size=0.055,
        y_title_size=0.055,
        label_size_x=0.045,
        label_size_y=0.045
    )

def draw_header(ptmin, ptmax, cut_name, extra=None, extra1 =None, x=0.16, size=0.06):
    latex = keep(root.TLatex())
    latex.SetNDC()
    latex.SetTextFont(42)

    latex.SetTextSize(size)
    latex.DrawLatex(x, 0.925, data_set+", "+run_text)

    latex.SetTextSize(size)
    latex.DrawLatex(x, 0.850, f"{ptmin:.1f} < p_{{T}}^{{V0}} < {ptmax:.1f} GeV/c")
    #latex.DrawLatex(x, 0.825, cut_name)

    if extra:
        latex.DrawLatex(x, 0.775, extra)

    if extra1:
        latex.DrawLatex(x, 0.700, extra1)

    return latex

def draw_zero_line(xmin, xmax):
    line = keep(root.TLine(xmin, 0., xmax, 0.))
    line.SetLineColor(root.kBlack)
    line.SetLineStyle(2)
    line.SetLineWidth(1)
    line.Draw("same")
    return line

def get_lambda_h2(cut_name, charge_name):
    if lambda_species == "Lambda":
        return get_hist(hist_path(cut_name, charge_name, "h_mass_Lambda_vs_v0_pt"))

    if lambda_species == "AntiLambda":
        return get_hist(hist_path(cut_name, charge_name, "h_mass_AntiLambda_vs_v0_pt"))

    h = get_hist(hist_path(cut_name, charge_name, "h_mass_Lambda_vs_v0_pt"))
    hbar = get_hist(hist_path(cut_name, charge_name, "h_mass_AntiLambda_vs_v0_pt"))
    h.Add(hbar)
    h.SetName(h.GetName()+"_LambdaPlusAntiLambda")
    return h
```

```python

def prepare_species_by_cut(
    particle,
    pt_bins,
    cuts_by_pt,
    rebins,
    sidebands,
    prefix,
    background_method="combined_like"
):

    if len(cuts_by_pt) != len(pt_bins)-1:
        raise RuntimeError(
            f"{prefix}: cuts list must have {len(pt_bins)-1} entries"
        )

    if len(rebins) != len(pt_bins)-1:
        raise RuntimeError(
            f"{prefix}: rebin list must have {len(pt_bins)-1} entries"
        )

    results = []

    for i in range(len(pt_bins)-1):

        ptmin = pt_bins[i]
        ptmax = pt_bins[i+1]

        cut_name = cuts_by_pt[i]
        rebin = rebins[i]

        if particle == "k0s":

            h2_unlike = get_hist(
                hist_path(
                    cut_name,
                    "unlike",
                    "h_mass_Kshort_vs_v0_pt"
                )
            )

            if background_method == "combined_like":

                h2_like = get_hist(
                    hist_path(
                        cut_name,
                        "like",
                        "h_mass_Kshort_vs_v0_pt"
                    )
                )

            elif background_method == "geometric":

                h2_pp = get_hist(
                    hist_path(
                        cut_name,
                        "plusplus",
                        "h_mass_Kshort_vs_v0_pt"
                    )
                )

                h2_mm = get_hist(
                    hist_path(
                        cut_name,
                        "minusminus",
                        "h_mass_Kshort_vs_v0_pt"
                    )
                )

        elif particle == "lambda":

            h2_unlike = get_lambda_h2(
                cut_name,
                "unlike"
            )

            if background_method == "combined_like":

                h2_like = get_lambda_h2(
                    cut_name,
                    "like"
                )

            elif background_method == "geometric":

                h2_pp = get_lambda_h2(
                    cut_name,
                    "plusplus"
                )

                h2_mm = get_lambda_h2(
                    cut_name,
                    "minusminus"
                )

        else:
            raise RuntimeError(
                "particle must be 'k0s' or 'lambda'"
            )

        h_fg = project_mass(
            h2_unlike,
            ptmin,
            ptmax,
            f"h_{prefix}_fg_pt{i}_{cut_name}"
        )

        if background_method == "combined_like":

            h_like_raw = project_mass(
                h2_like,
                ptmin,
                ptmax,
                f"h_{prefix}_like_pt{i}_{cut_name}"
            )

        elif background_method == "geometric":

            h_pp = project_mass(
                h2_pp,
                ptmin,
                ptmax,
                f"h_{prefix}_pp_pt{i}_{cut_name}"
            )

            h_mm = project_mass(
                h2_mm,
                ptmin,
                ptmax,
                f"h_{prefix}_mm_pt{i}_{cut_name}"
            )

        if rebin > 1:

            h_fg.Rebin(rebin)

            if background_method == "combined_like":
                h_like_raw.Rebin(rebin)

            elif background_method == "geometric":
                h_pp.Rebin(rebin)
                h_mm.Rebin(rebin)

        if background_method == "geometric":

            h_like_raw = make_geometric_like_background(
                h_pp,
                h_mm,
                f"h_{prefix}_geometric_pt{i}_{cut_name}"
            )

        h_bg, scale, scale_err = scale_like_to_unlike(
            h_fg,
            h_like_raw,
            sidebands
        )

        h_sub = make_subtracted(
            h_fg,
            h_bg,
            f"h_{prefix}_sub_pt{i}_{cut_name}"
        )

        results.append({
            "ptmin": ptmin,
            "ptmax": ptmax,
            "cut": cut_name,
            "rebin": rebin,
            "fg": h_fg,
            "like_raw": h_like_raw,
            "bg": h_bg,
            "sub": h_sub,
            "scale": scale,
            "scale_err": scale_err,
            "background_method": background_method
        })

    return results
```

```python

def fit_gaus_then_gaus_pol2(h, name, gaus_range, fit_range,
                             mean_limits, sigma_limits):

    gaus = keep(root.TF1(
        name+"_gaus",
        "gaus",
        gaus_range[0],
        gaus_range[1]
    ))

    peak_bin = h.GetMaximumBin()
    amp0 = max(h.GetBinContent(peak_bin), 1.)
    mean0 = h.GetBinCenter(peak_bin)
    sigma0 = 0.010

    gaus.SetParameters(amp0, mean0, sigma0)
    gaus.SetParLimits(1, mean_limits[0], mean_limits[1])
    gaus.SetParLimits(2, sigma_limits[0], sigma_limits[1])

    h.Fit(gaus, "RQ0")

    full = keep(root.TF1(
        name+"_gaus_pol2",
        "gaus(0)+pol2(3)",
        fit_range[0],
        fit_range[1]
    ))

    full.SetParameters(
        gaus.GetParameter(0),
        gaus.GetParameter(1),
        abs(gaus.GetParameter(2)),
        0., 0., 0.
    )

    full.SetParLimits(1, mean_limits[0], mean_limits[1])
    full.SetParLimits(2, sigma_limits[0], sigma_limits[1])

    h.Fit(full, "RQ0")

    signal = keep(root.TF1(
        name+"_signal",
        "gaus",
        fit_range[0],
        fit_range[1]
    ))

    signal.SetParameters(
        full.GetParameter(0),
        full.GetParameter(1),
        full.GetParameter(2)
    )

    background = keep(root.TF1(
        name+"_background",
        "pol2",
        fit_range[0],
        fit_range[1]
    ))

    background.SetParameters(
        full.GetParameter(3),
        full.GetParameter(4),
        full.GetParameter(5)
    )

    return {
        "gaus": gaus,
        "full": full,
        "signal": signal,
        "background": background
    }
```

```python
nrows_k0s = math.ceil((len(pt_k0s)-1)/ncols)
nrows_lambda = math.ceil((len(pt_lambda)-1)/ncols)

k0s_results = prepare_species_by_cut(
    particle="k0s",
    pt_bins=pt_k0s,
    cuts_by_pt=cuts_k0s,
    rebins=rebin_k0s,
    sidebands=sidebands_k0s,
    prefix="k0s",
    background_method=like_background_method
)

lambda_results = prepare_species_by_cut(
    particle="lambda",
    pt_bins=pt_lambda,
    cuts_by_pt=cuts_lambda,
    rebins=rebin_lambda,
    sidebands=sidebands_lambda,
    prefix="lambda",
    background_method=like_background_method
)
```

```python
#K0S foreground and scaled like-sign background
#fit the unlike-sign foreground with Gaussian + pol2

k0s_fg_fits = []

c_k0s_fg_bg_fit = root.TCanvas(
    "c_k0s_fg_bg_fit",
    "c_k0s_fg_bg_fit",
    1500,
    450*nrows_k0s
)

c_k0s_fg_bg_fit.Divide(ncols, nrows_k0s)

legends_k0s_fg_bg_fit = []

for i, result in enumerate(k0s_results):

    c_k0s_fg_bg_fit.cd(i+1)

    root.gPad.SetLeftMargin(0.14)
    root.gPad.SetBottomMargin(0.14)
    root.gPad.SetRightMargin(0.01)
    root.gPad.SetTopMargin(0.01)

    h_fg = result["fg"]
    h_bg = result["bg"]

    #first Gaussian fit in the narrow K0S peak region
    gaus_fg = keep(root.TF1(
        f"gaus_k0s_fg_pt{i}",
        "gaus",
        gaus_range_k0s[0],
        gaus_range_k0s[1]
    ))

    peak_bin = h_fg.GetMaximumBin()
    amplitude0 = max(h_fg.GetBinContent(peak_bin), 1.)
    mean0 = h_fg.GetBinCenter(peak_bin)
    sigma0 = 0.010

    gaus_fg.SetParameters(
        amplitude0,
        mean0,
        sigma0
    )

    gaus_fg.SetParLimits(
        1,
        mean_limits_k0s[0],
        mean_limits_k0s[1]
    )

    gaus_fg.SetParLimits(
        2,
        sigma_limits_k0s[0],
        sigma_limits_k0s[1]
    )

    h_fg.Fit(
        gaus_fg,
        "RQ0"
    )

    #then Gaussian + pol2 in the wider fit region
    full_fg = keep(root.TF1(
        f"gaus_pol2_k0s_fg_pt{i}",
        "gaus(0)+pol2(3)",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    full_fg.SetParameters(
        gaus_fg.GetParameter(0),
        gaus_fg.GetParameter(1),
        abs(gaus_fg.GetParameter(2)),
        0.,
        0.,
        0.
    )

    full_fg.SetParLimits(
        1,
        mean_limits_k0s[0],
        mean_limits_k0s[1]
    )

    full_fg.SetParLimits(
        2,
        sigma_limits_k0s[0],
        sigma_limits_k0s[1]
    )

    h_fg.Fit(
        full_fg,
        "RQ0"
    )

    signal_fg = keep(root.TF1(
        f"signal_k0s_fg_pt{i}",
        "gaus",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    signal_fg.SetParameters(
        full_fg.GetParameter(0),
        full_fg.GetParameter(1),
        full_fg.GetParameter(2)
    )

    background_fg = keep(root.TF1(
        f"background_k0s_fg_pt{i}",
        "pol2",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    background_fg.SetParameters(
        full_fg.GetParameter(3),
        full_fg.GetParameter(4),
        full_fg.GetParameter(5)
    )

    k0s_fg_fits.append({
        "gaus": gaus_fg,
        "full": full_fg,
        "signal": signal_fg,
        "background": background_fg
    })

    #automatic Y-axis scaling
    ymax_raw = max(
        h_fg.GetMaximum(),
        h_bg.GetMaximum()
    )

    y_power = max(0, int(math.floor(math.log10(ymax_raw))))
    y_scale = 10.**y_power

    h_fg.Scale(1./y_scale)
    h_bg.Scale(1./y_scale)

    full_fg.SetParameter(0, full_fg.GetParameter(0)/y_scale)
    full_fg.SetParameter(3, full_fg.GetParameter(3)/y_scale)
    full_fg.SetParameter(4, full_fg.GetParameter(4)/y_scale)
    full_fg.SetParameter(5, full_fg.GetParameter(5)/y_scale)

    signal_fg.SetParameter(0, signal_fg.GetParameter(0)/y_scale)

    background_fg.SetParameter(0, background_fg.GetParameter(0)/y_scale)
    background_fg.SetParameter(1, background_fg.GetParameter(1)/y_scale)
    background_fg.SetParameter(2, background_fg.GetParameter(2)/y_scale)

    setup_mass_hist(
        h_fg,
        "m_{#pi^{+}#pi^{-}} [GeV/c^{2}]",
        "pairs [{:.0f}]".format(y_scale)
    )

    h_fg.GetXaxis().SetRangeUser(
        mass_range_k0s[0],
        mass_range_k0s[1]
    )

    h_fg.SetLineColor(root.kBlack)
    h_fg.SetMarkerColor(root.kBlack)
    h_fg.SetMarkerStyle(20)
    h_fg.SetMarkerSize(0.7)
    h_fg.SetLineWidth(2)

    h_bg.SetLineColor(root.kRed+1)
    h_bg.SetMarkerColor(root.kRed+1)
    h_bg.SetMarkerStyle(24)
    h_bg.SetMarkerSize(0.7)
    h_bg.SetLineWidth(2)

    ymax = max(
        h_fg.GetMaximum(),
        h_bg.GetMaximum()
    )

    h_fg.SetMinimum(0.)
    h_fg.SetMaximum(1.45*ymax)

    Format_hist_loc(
        h_fg,
        title="",
        x_title="m_{#pi^{+}#pi^{-}} [GeV/c^{2}]",
        y_title=f"pairs [{y_scale:.0f}]",
        x_title_offset=1.0,
        y_title_offset=0.99,
        center_title=True,
        x_title_size=0.06,
        y_title_size=0.06,
        label_size_x=0.06,
        label_size_y=0.06
    )

    h_fg.GetXaxis().SetNdivisions(505)
    h_fg.GetYaxis().SetNdivisions(505)
    h_fg.GetYaxis().SetNoExponent(True)
    h_fg.GetYaxis().SetMaxDigits(2)

    h_fg.Draw("E")
    h_bg.Draw("E same")

    full_fg.SetLineColor(root.kBlue+1)
    full_fg.SetLineWidth(2)
    full_fg.Draw("same")

    signal_fg.SetLineColor(root.kMagenta+1)
    signal_fg.SetLineWidth(2)
    signal_fg.SetLineStyle(2)
    signal_fg.Draw("same")

    background_fg.SetLineColor(root.kGreen+2)
    background_fg.SetLineWidth(2)
    background_fg.SetLineStyle(3)
    background_fg.Draw("same")

    legend = keep(root.TLegend(
        0.67,
        0.6,
        0.98,
        0.985
    ))

    legend.SetBorderSize(0)
    legend.SetFillStyle(0)
    legend.SetTextFont(42)
    legend.SetTextSize(0.05)

    legend.AddEntry(
        h_fg,
        "unlike sign",
        "lep"
    )

    legend.AddEntry(
        h_bg,
        "scaled like",
        "lep"
    )

    legend.AddEntry(
        full_fg,
        "Gaus + pol2",
        "l"
    )

    legend.AddEntry(
        signal_fg,
        "Gauss",
        "l"
    )

    legend.AddEntry(
        background_fg,
        "pol2",
        "l"
    )

    legend.Draw()
    legends_k0s_fg_bg_fit.append(legend)

    mean = full_fg.GetParameter(1)
    mean_error = full_fg.GetParError(1)

    sigma = abs(
        full_fg.GetParameter(2)
    )

    sigma_error = full_fg.GetParError(2)

    draw_header(
        result["ptmin"],
        result["ptmax"],
        result["cut"],
        f"#mu = {mean:.4f} #pm {mean_error:.4f}",
        f"#sigma = {sigma:.4f} #pm {sigma_error:.4f}",
        size=0.05
    )

c_k0s_fg_bg_fit.Draw()

if True:
    c_k0s_fg_bg_fit.SaveAs(
        outdir +
        "/K0S_fg_scaledLS_with_fit_variable_cuts.pdf"
    )
```

```python
#K0S foreground and scaled like-sign background
#fit the unlike-sign foreground with Gaussian + pol2

k0s_fg_fits = []

c_k0s_fg_bg_fit = root.TCanvas(
    "c_k0s_fg_bg_fit",
    "c_k0s_fg_bg_fit",
    1500,
    450*nrows_k0s
)

c_k0s_fg_bg_fit.Divide(ncols, nrows_k0s)

legends_k0s_fg_bg_fit = []

for i, result in enumerate(k0s_results):

    c_k0s_fg_bg_fit.cd(i+1)

    root.gPad.SetLeftMargin(0.10)
    root.gPad.SetBottomMargin(0.13)
    root.gPad.SetRightMargin(0.02)
    root.gPad.SetTopMargin(0.07)

    h_fg = result["fg"]
    h_bg = result["bg"]

    #first Gaussian fit in the narrow K0S peak region
    gaus_fg = keep(root.TF1(
        f"gaus_k0s_fg_pt{i}",
        "gaus",
        gaus_range_k0s[0],
        gaus_range_k0s[1]
    ))

    peak_bin = h_fg.GetMaximumBin()
    amplitude0 = max(h_fg.GetBinContent(peak_bin), 1.)
    mean0 = h_fg.GetBinCenter(peak_bin)
    sigma0 = 0.010

    gaus_fg.SetParameters(
        amplitude0,
        mean0,
        sigma0
    )

    gaus_fg.SetParLimits(
        1,
        mean_limits_k0s[0],
        mean_limits_k0s[1]
    )

    gaus_fg.SetParLimits(
        2,
        sigma_limits_k0s[0],
        sigma_limits_k0s[1]
    )

    h_fg.Fit(
        gaus_fg,
        "RQ0"
    )

    #then Gaussian + pol2 in the wider fit region
    full_fg = keep(root.TF1(
        f"gaus_pol2_k0s_fg_pt{i}",
        "gaus(0)+pol2(3)",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    full_fg.SetParameters(
        gaus_fg.GetParameter(0),
        gaus_fg.GetParameter(1),
        abs(gaus_fg.GetParameter(2)),
        0.,
        0.,
        0.
    )

    full_fg.SetParLimits(
        1,
        mean_limits_k0s[0],
        mean_limits_k0s[1]
    )

    full_fg.SetParLimits(
        2,
        sigma_limits_k0s[0],
        sigma_limits_k0s[1]
    )

    h_fg.Fit(
        full_fg,
        "RQ0"
    )

    signal_fg = keep(root.TF1(
        f"signal_k0s_fg_pt{i}",
        "gaus",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    signal_fg.SetParameters(
        full_fg.GetParameter(0),
        full_fg.GetParameter(1),
        full_fg.GetParameter(2)
    )

    background_fg = keep(root.TF1(
        f"background_k0s_fg_pt{i}",
        "pol2",
        fit_range_k0s[0],
        fit_range_k0s[1]
    ))

    background_fg.SetParameters(
        full_fg.GetParameter(3),
        full_fg.GetParameter(4),
        full_fg.GetParameter(5)
    )

    k0s_fg_fits.append({
        "gaus": gaus_fg,
        "full": full_fg,
        "signal": signal_fg,
        "background": background_fg
    })

    setup_mass_hist(
        h_fg,
        "m_{#pi^{+}#pi^{-}} [GeV/c^{2}]",
        "pairs"
    )

    h_fg.GetXaxis().SetRangeUser(
        mass_range_k0s[0],
        mass_range_k0s[1]
    )

    h_fg.SetLineColor(root.kBlack)
    h_fg.SetMarkerColor(root.kBlack)
    h_fg.SetMarkerStyle(20)
    h_fg.SetMarkerSize(0.7)
    h_fg.SetLineWidth(2)

    h_bg.SetLineColor(root.kRed+1)
    h_bg.SetMarkerColor(root.kRed+1)
    h_bg.SetMarkerStyle(24)
    h_bg.SetMarkerSize(0.7)
    h_bg.SetLineWidth(2)

    ymax = max(
        h_fg.GetMaximum(),
        h_bg.GetMaximum()
    )

    h_fg.SetMinimum(0.)
    h_fg.SetMaximum(1.45*ymax)

    Format_hist_loc(h_fg,      
        title="",
        x_title="m_{#pi^{+}#pi^{-}} [GeV/c^{2}]",
        y_title="pairs",
        x_title_offset=1.0,
        y_title_offset=0.8,
        center_title=True,
        x_title_size=0.06,
        y_title_size=0.06,
        label_size_x=0.06,
        label_size_y=0.06
    )
    
    h_fg.GetXaxis().SetNdivisions(505)
    h_fg.GetYaxis().SetNdivisions(505)
    h_fg.GetYaxis().SetMaxDigits(3)

    h_fg.Draw("E")
    h_bg.Draw("E same")

    full_fg.SetLineColor(root.kBlue+1)
    full_fg.SetLineWidth(2)
    full_fg.Draw("same")

    signal_fg.SetLineColor(root.kMagenta+1)
    signal_fg.SetLineWidth(2)
    signal_fg.SetLineStyle(2)
    signal_fg.Draw("same")

    background_fg.SetLineColor(root.kGreen+2)
    background_fg.SetLineWidth(2)
    background_fg.SetLineStyle(3)
    background_fg.Draw("same")

    legend = keep(root.TLegend(
        0.62,
        0.57,
        0.95,
        0.96
    ))

    legend.SetBorderSize(0)
    legend.SetFillStyle(0)
    legend.SetTextFont(42)
    legend.SetTextSize(0.05)

    legend.AddEntry(
        h_fg,
        "unlike sign",
        "lep"
    )

    legend.AddEntry(
        h_bg,
        "scaled like",
        "lep"
    )

    legend.AddEntry(
        full_fg,
        "Gaus + pol2",
        "l"
    )

    legend.AddEntry(
        signal_fg,
        "Gauss",
        "l"
    )

    legend.AddEntry(
        background_fg,
        "pol2",
        "l"
    )

    legend.Draw()
    legends_k0s_fg_bg_fit.append(legend)

    mean = full_fg.GetParameter(1)
    mean_error = full_fg.GetParError(1)

    sigma = abs(
        full_fg.GetParameter(2)
    )

    sigma_error = full_fg.GetParError(2)

    draw_header(
        result["ptmin"],
        result["ptmax"],
        cuts_k0s[0],
        f"#mu = {mean:.4f} #pm {mean_error:.4f}",
        f"#sigma = {sigma:.4f} #pm {sigma_error:.4f}",
        size=0.05
        
    )

c_k0s_fg_bg_fit.Draw()

if True:
    c_k0s_fg_bg_fit.SaveAs(
        outdir +
        f"/K0S_fg_scaledLS_with_fit_{cuts_k0s[0]}.pdf"
    )
```
