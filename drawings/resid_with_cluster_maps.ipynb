```python
def Format_hist_loc(h, title, x_title, y_title, x_title_size=0.06, y_title_size=0.06, x_title_offset=1.0, \
                    y_title_offset=1.0, label_size_x=0.06, label_size_y=0.06, label_size_z=0.06,\
                         z_axis_offset=1.0, z_range = None, center_title=False, title_size=0.06):
    if h is None:
        print("Error: Histogram is None. Cannot format.")
        return
    if title is not None:
        h.SetTitle(title)
    if x_title is not None:
        h.GetXaxis().SetTitle(x_title)
    if y_title is not None:
        h.GetYaxis().SetTitle(y_title)
    h.GetXaxis().SetTitleSize(x_title_size)
    h.GetYaxis().SetTitleSize(y_title_size)
    h.GetXaxis().SetTitleOffset(x_title_offset)
    h.GetYaxis().SetTitleOffset(y_title_offset)
    h.GetXaxis().SetLabelSize(label_size_x)
    h.GetYaxis().SetLabelSize(label_size_y)
    h.GetZaxis().SetLabelSize(label_size_z)
    h.GetZaxis().SetTitleOffset(z_axis_offset)
    #h.SetTitleSize(title_size)
    if z_range is not None:
        h.GetZaxis().SetRangeUser(z_range[0], z_range[1])
    if center_title:
        h.GetXaxis().CenterTitle()
        h.GetYaxis().CenterTitle()
```

```python

#helper functions for layer/phi and signed-z/layer residual maps
#keeps the same ROOT drawing style used above

def Draw_layer_lines(xmin, xmax, ymin, ymax, x_is_layer=False, y_is_layer=False):
    lines = []
    for layer_edge in [7, 23, 39, 55]:
        if x_is_layer:
            line = root.TLine(layer_edge, ymin, layer_edge, ymax)
            line.SetLineColor(root.kBlack)
            line.SetLineWidth(1)
            line.SetLineStyle(2)
            line.Draw("same")
            lines.append(line)
        if y_is_layer:
            line = root.TLine(xmin, layer_edge, xmax, layer_edge)
            line.SetLineColor(root.kBlack)
            line.SetLineWidth(1)
            line.SetLineStyle(2)
            line.Draw("same")
            lines.append(line)
    return lines

def Draw_phi_lines(xmin, xmax):
    lines = []
    sector_angle = root.TMath.Pi()/6.
    angle_start = root.TMath.Pi() - sector_angle/2.
    for isector in range(12):
        phi_edge = angle_start - isector*sector_angle
        if phi_edge > root.TMath.Pi():
            phi_edge -= 2.*root.TMath.Pi()
        if phi_edge < -root.TMath.Pi():
            phi_edge += 2.*root.TMath.Pi()
        line = root.TLine(xmin, phi_edge, xmax, phi_edge)
        line.SetLineColor(root.kBlack)
        line.SetLineWidth(1)
        line.SetLineStyle(2)
        line.Draw("same")
        lines.append(line)
    return lines

def Draw_z_lines(ymin, ymax):
    lines = []
    for z_edge in [-102.325, 0., 102.325]:
        line = root.TLine(z_edge, ymin, z_edge, ymax)
        line.SetLineColor(root.kBlack)
        line.SetLineWidth(1)
        line.SetLineStyle(2)
        line.Draw("same")
        lines.append(line)
    return lines

def Format_residual_map(h, title, x_title, y_title, z_title, x_range, y_range, z_range):
    Format_hist_loc(h, title=title, x_title=x_title, y_title=y_title,
                    x_title_offset=.9, y_title_offset=1.0,
                    center_title=True, x_title_size=0.06,
                    y_title_size=0.06, label_size_x=0.05,
                    label_size_y=0.05, label_size_z=0.05,
                    z_axis_offset=1.15, z_range=z_range)
    h.GetZaxis().SetTitle(z_title)
    h.GetXaxis().SetRangeUser(x_range[0], x_range[1])
    h.GetYaxis().SetRangeUser(y_range[0], y_range[1])
```
