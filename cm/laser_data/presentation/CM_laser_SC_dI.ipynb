# Space charge from ΔI_GEM inside the 795xx fill (truth-free)
11 beam blocks of 79508 / 79513 / 79516 / 79529 (I_S 1.81–2.74, I_N 1.93–2.87 µA), one fill, same static and alignment.
For every CM stripe the corrected R dφ is fitted linearly vs the block GEM current: the **slope** is
- with the static-only correction: the **space-charge response** of the data per µA;
- with a full correction (CDB, or PHGarfield map × b): the part of the response the model **misses**.

The CM truth, the CM → pad alignment and the static model are constant over the fill and cancel. Sector edges excluded.

```python
%matplotlib inline
import numpy as np, matplotlib.pyplot as plt
import truthalign as ta, scdiff as sd, slidestyle
plt.rcParams['figure.dpi'] = 80
plt.rcParams.update({'font.size': 14, 'axes.titlesize': 15, 'axes.labelsize': 15, 'legend.fontsize': 11})
def show(*f):
    for x in f: display(slidestyle.restyle(x))
    plt.close('all')
ta.correct_truth()
```

## 1. Response and residuals vs R

```python
show(sd.fig_response())
```

## 2. b scan per map: R² residual and rms

```python
show(sd.fig_bscan())
```

**Reading.**
- Data: d(RΔφ)/dI ≈ +1.4 → −1.5 mm/µA from 31 to 73 cm, R² ≈ 865 µm/µA, the same on both sides.
- The CDB space-charge part matches the data R² per µA (S +3, N −46 µm/µA).
- Only **α 1.8** has its R² zero where the rms is minimal on both sides: **b ≈ 0.67 (S) / 0.59 (N)**. With α 1.6 the
  curvature wants b ≈ 0.73 / 0.64 but the shape b ≈ 0.65 / 0.60: α 1.6 leaves an R shape.
- Beyond the quadratic, every model leaves 0.04–0.09 mm/µA (< 0.2 mm at 79513): the residual "sine in R" is small.

## 3. φ dependence of the response

```python
show(sd.fig_sectors())
```

**Reading.** Sector-to-sector scatter of the response is ≈ 300–450 µm/µA, with a South first harmonic at ≈ −150°: the
opposite phase of the beam − no-beam sine (+70°), which therefore was a time-dependent static / alignment difference,
not space charge. Statistical precision per sector still to be checked (split samples).
