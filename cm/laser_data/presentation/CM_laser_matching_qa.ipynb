# CM laser: matching QA

Two kinds of matching are used in the study:
1. **to the CM truth pattern** (LaserClusterTruthMatcher): each cluster is assigned to a truth stripe; the residual is cluster - stripe.
   Works for the corrected beam runs; fails (rows shifted by one row spacing) for uncorrected clusters and for the no-beam /
   field-off runs, whose distortions w.r.t. the pattern are larger than half a row spacing in places.
2. **cluster to cluster** between two runs (mutual nearest neighbours within 1 cm): beam - no beam and field on - field off;
   for field off the radial offset per R ring (several mm) is removed first and the window is 6 mm ("robust").
Data: final configuration of all 17 blocks; `data/match/` (ExtractResiduals.C, modes truthqa / c2cqa / robustqa).

Margin = distance to the second-nearest / nearest reference cluster (large = unambiguous); matched fraction = mutual nearest
