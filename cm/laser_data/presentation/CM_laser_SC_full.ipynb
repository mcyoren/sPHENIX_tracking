# CM laser: full space-charge study – what the corrections leave of the space-charge change between runs (2026-10-23, v2)

**Idea.** The difference of two runs removes the CM truth and the alignment, and inside one fill also the static. What is left
is the change of the space-charge distortion with the current (about 2–3 mm R² and 20–30 mm pointing between fills, Section 1
of v1). With a correction applied, the difference shows what that correction **leaves** of the change: 0 = perfect.

**Runs**: 79508 / 79513 / 79516 (one fill, 1.8–2.9 µA), 80892 (5.4–6.3 µA), 81558 (6.1–7.0 µA); a run = mean of its blocks.
Grey columns in the observable plots: pairs inside one fill (no static change). **Corrections** (same clusters):
current CDB (red), final maps f9_sper (blue: South α 1.7 mix / North α 1.6, keff = a I − 0.015 I², per-period static), final +
CM module weights f13_gain (green).

**Profile**: dφ(R) = K0s-weighted median over φ of corrected cluster φ − aligned truth φ in 2 cm R bins (standard pipeline:
sector edges ±6° excluded, own matching to the truth, field-off map subtracted). **Observables** of a profile:
arch (R1 minus a line fitted to R > 42 cm, mrad), sideways shift (constant R·dφ, mm), R² (quadratic coefficient of a fit to
the R bins as a sagitta over 46 cm, µm), pointing (that fit at R = 0, mm), and **what a TPC-only track sees**: a + bR + cR²
fitted to R·dφ at the 48 pad-layer radii with equal weight per layer (R1 has a third of the layers), sagitta c·L²/4 over
L = 48 cm (µm) and offset at R = 0 (mm). At 1 GeV and 1.4 T a 100 µm sagitta is a 0.8 % momentum bias (K0s daughters:
opposite charges, ± the same bias).

**Error bars** in Section 2 are not statistical: they span the values of the same observable for five profile definitions
(standard R > 32 cm, standard R > 34 cm, all φ with truth-R bins, edge cut with truth-R bins, track fit without the first two
layers). The statistical error is much smaller.

**Findings**
1. The run-to-run dφ per R bin is measured robustly, including the R1 arch of the CDB that you questioned. In 79516 − 81558
   (CDB), R1 is −1.3 … −2.4 mrad (South) and −2.3 … −3.6 mrad (North) with every estimator, stripe set, weighting and edge
   treatment (Section 6a).
2. The track-relevant numbers derived from it are **not** robust: they depend on the innermost R1 bins (30–34 cm), where the
   profile is steep at high current. R² and pointing change sign with the inner R cut (Section 6b), and the track sagitta
   moves by 150–400 µm when the first two layers are dropped (Section 6c).
3. So the CM cannot yet say what the tracks see at better than about ±200 µm sagitta at high current, and the K0s are the
   arbiter. A CDB R1 arch is no contradiction with symmetric K0s daughters if it sits in the first layers. Whether those bins
   are real distortion or a CM artifact at the inner field cage is the open question (Section 7).

### 6b. The derived observables vs the inner R cut and the profile definition
R² and pointing of 79516 − 81558 for three definitions (mine all φ | mine edge cut | standard) with the quadratic fit starting at
32, 34, 36, 42 cm.

```python
V = {}
import cmpos as cm
for tag in ('cdb', 'f9_sper', 'f13_gain'):
    T = cm.use(None if tag == 'f9_sper' else tag)
    for s in (0, 1):
        V[(tag, s, 0)] = rc.run_prof(T, '79516', s, 'mod', None, True) - rc.run_prof(T, '81558', s, 'mod', None, True)
        V[(tag, s, 1)] = rc.run_prof(T, '79516', s, 'mod', None, True, 6.0) - rc.run_prof(T, '81558', s, 'mod', None, True, 6.0)
        V[(tag, s, 2)] = sq.run_prof(tag, s, '79516') - sq.run_prof(tag, s, '81558')
cm.use(None)
old = sq.RMIN
for rmin in (32.0, 34.0, 36.0, 42.0):
    sq.RMIN = rmin
    print(f'R > {rmin:.0f} cm: R2 [um] / pointing [mm]  (mine all phi | mine edge cut | standard)')
    for tag in ('cdb', 'f9_sper', 'f13_gain'):
        for s in (0, 1):
            ms = [sq.metrics(V[(tag, s, k)]) for k in range(3)]
            print(f"   {tag:9s} {'SN'[s]}: " + ' | '.join(f"{m['R2']:+5.0f} {m['pointing']:+5.2f}" for m in ms))
sq.RMIN = old
```

## 8. Summary and the test that decides
* Per R bin the CM measures the run-to-run dφ robustly (also the CDB R1 arch: same with every estimator, stripe set and
  weighting).
* The track-relevant numbers (sagitta, pointing) of the φ-averaged profile are dominated by the innermost R1 bins (30–34 cm) and
  change by 150–400 µm with their treatment, so the CM alone does not fix what tracks see at high current to better than about
  ±200 µm.
* **Decisive test with the K0s**: per run and correction (CDB, final), the measured K0s daughter balance / mass vs the CM track
  prediction of Section 6c (all layers vs without the first two). The definition that predicts the K0s tells whether the inner R1
  bins are real; then the space-charge and static fits use that definition.
* Per-module space charge (Section 4 and CM_laser_cluster_position): the module weights reduce the 30° residuals by 15–25 %
  and leave more than the noise.
