# CM laser: per-module static and space charge – towards ≤ 200 µm track sagitta in every sector (2026-10-23)

**Target**: the sagitta a TPC-only track sees (fit a + bR + cR² to R·dφ at the 48 pad-layer radii, equal weight per layer,
sagitta c·L²/4 over 48 cm) ≤ 200 µm in every 30° sector. Measured per sector from the CM residual (corrected cluster φ − truth φ,
3-pad COG 2 rows, final configuration f9_sper) in cells of ¼ module (sector × ring × φ half × R half → 6 radial points per sector).

**Model** (per side, all 14 beam blocks of 79508 / 79513 / 79516 / 80892 / 81558): cell residual
Y_b(c) = A(c) + g(c)·I_b + per fill (80892, 81558) ring means + translation.
A = static left at I = 0 (what the static model misses per cell), g = space-charge slope left per cell (what the maps miss), the
fill terms = φ-symmetric period static (field cage / frames). A sagitta needs an R dependence inside a sector, so the ring means
(constant per ring) are kept in A; they are removed only in the comparison plots (global level).

**Results**
* As measured, the sector sagitta is 1.2 / 0.83 mm rms (South / North); 11 / 22 % of sectors within ±200 µm.
* The per-cell static alone removes little (0.89 / 0.68 mm): at high current the per-cell space charge dominates.
* Per-cell static + linear space charge: **135 / 90 µm rms, 93 / 98 % within ±200 µm** (in-sample fit, 2 numbers per cell).
* Leave-one-run-out (fit without the run, predict it; only its 5 φ-symmetric numbers fitted): 795xx 55–130 µm rms (91–100 %
  within), 80892 / 81558 (a whole fill predicted from the others) 165–293 µm rms (56–83 %); all runs 192 / 144 µm rms, 83 / 90 %.
  The misses at high current: the per-cell space charge is not exactly linear in the current.
* Leave one **block** out (production situation, the rest of the fill known): **100–106 µm rms, 93–94 % of sectors within
  ±200 µm** on both sides (worst 0.4–0.5 mm, at high current).
* The low-R kink (v14): keff 4–8 % lower at high current removes it (keff = a I − c I², c ≈ 0.019 instead of 0.015).
* The static sector pattern at I = 0 is the one of the no-beam run 75073 (sector-sagitta correlation +0.90 South / +0.81 North;
  83319 North +0.84).

## 4. (b) Per-module frame charge as the physical static knob
The frame field acts within a few cm of the frames, so the response of a cell to its own module's frame charge is the global
frame response (c8 ±30 V, all 14 blocks) restricted to the module. Fit of one frame-voltage change per module to the
sector-dependent static: the explained part is small and the voltages are unphysical (hundreds of V to 2 kV, frames are
100–180 V) – the per-module static is not frame charge. It correlates with the no-beam field-on run, not with field off: an E×B
effect of per-module radial fields (readout / GEM plane), for which PHGarfield has no ingredient yet.

## 6. Summary and next steps
* With a per-cell static and a per-cell linear space-charge term the CM reaches the target in-sample (93 / 98 % of sectors within
  ±200 µm) and nearly out of sample at low current; at high current a non-linear term per cell is needed (sublinear growth, as
  the global keff).
* The static per cell agrees with the no-beam run: it can be taken from the no-beam runs, the beam fit at I = 0, or both.
* Implementation options: (1) a data-driven CM residual map (sector × R per period: static + slope × I), applied on top of the
  model correction – what the production CM distortion maps do, but with the current dependence and the period static from the
  fit; (2) the physical route: per-module frame / gain weights in PHGarfield (v13 gains + per-module frame charge).
* Caveat: the sagitta uses the inner R1 cells (30–40 cm) as all CM numbers; their reliability is the open K0s question.
